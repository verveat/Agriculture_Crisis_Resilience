```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ============================================================
# 1. TIME RANGE AND DATE HELPERS
# ============================================================

first_year = int(production_plot["Startjahr"].min())
last_year = int(production_plot["Startjahr"].max())
plot_end_year = last_year + 1

tick_years = np.arange(first_year, last_year + 1)
marketing_year_labels = [
    f"{year}/{year + 1}" for year in tick_years
]

data_start = pd.Timestamp(f"{first_year}-07-01")
data_end_exclusive = pd.Timestamp(f"{last_year + 1}-07-01")


def to_timestamp(value):
    """Convert a date-like value to Timestamp; return None if absent."""
    if value is None or pd.isna(value) or str(value).strip() == "":
        return None
    return pd.Timestamp(value)


def financial_year_x(date):
    """
    Map a calendar date onto the continuous financial-year axis.
    July 1 = x.0; the following July 1 = x+1.0.
    """
    date = pd.Timestamp(date)
    fy_start_year = date.year if date.month >= 7 else date.year - 1
    fy_start = pd.Timestamp(f"{fy_start_year}-07-01")
    fy_end = pd.Timestamp(f"{fy_start_year + 1}-07-01")

    return fy_start_year + (
        (date - fy_start).total_seconds()
        / (fy_end - fy_start).total_seconds()
    )


def clip_interval(start, end):
    """Clip an interval to the observed BLE sample."""
    start = max(start, data_start)
    end = min(end, data_end_exclusive)

    if start >= end:
        return None

    return start, end


# ============================================================
# 2. PRODUCTION LINE STYLES
# ============================================================

product_styles = {
    "Weichweizen": {
        "color": "steelblue",
        "marker": "o",
        "label": "Soft wheat",
    },
    "Gerste": {
        "color": "darkorange",
        "marker": "s",
        "label": "Barley",
    },
    "Kartoffeln": {
        "color": "forestgreen",
        "marker": "^",
        "label": "Potatoes",
    },
}

event_color = "#8C8C8C"


# ============================================================
# 3. PREPARE EVENT WINDOWS
# ============================================================

prepared_events = []

for event in events:
    # The analysis window defines the research period.
    # Actual event dates are retained separately for reference.
    analysis_start = to_timestamp(event.get("analysis_start_date"))
    analysis_end = to_timestamp(event.get("analysis_end_date"))

    actual_start = to_timestamp(event.get("event_start_date"))
    actual_end = to_timestamp(event.get("event_end_date"))

    # Prefer analysis boundaries; fall back to actual event dates.
    start = analysis_start or actual_start
    end = analysis_end or actual_end

    if start is None:
        print(f"Skipping {event.get('id')}: no start date available.")
        continue

    # Open-ended analysis windows extend only to the observed sample end.
    if end is None:
        end = data_end_exclusive

    clipped = clip_interval(start, end)

    prepared_events.append({
        **event,
        "_start": start,
        "_end": end,
        "_actual_start": actual_start,
        "_actual_end": actual_end,
        "_clipped": clipped,
    })

# Keep a chronological order
prepared_events.sort(key=lambda e: e["_start"])


# ============================================================
# FIGURE 1: PRODUCTION AND MONTHLY EVENT WINDOWS
# ============================================================

fig, ax = plt.subplots(figsize=(17, 8))

# Event windows use exact dates, not whole-year approximations.
# Neutral shading avoids implying that a risk category's effect
# on production has already been demonstrated.
for event in prepared_events:
    interval = event["_clipped"]

    if interval is None:
        continue

    start, end = interval
    x_start = financial_year_x(start)
    x_end = financial_year_x(end)

    ax.axvspan(
        x_start,
        x_end,
        color=event_color,
        alpha=0.13,
        linewidth=0,
        zorder=0,
    )

    # Add a vertical marker for a documented event start date.
    actual_start = event["_actual_start"]

    if actual_start is not None and data_start <= actual_start < data_end_exclusive:
        ax.axvline(
            financial_year_x(actual_start),
            color="#555555",
            linewidth=0.8,
            linestyle=":",
            alpha=0.75,
            zorder=1,
        )


# Production lines: retain the +0.5 financial-year alignment
for product, style in product_styles.items():
    subset = production_plot[
        production_plot["Bilanz Produkt"] == product
    ].copy()

    subset = subset.dropna(
        subset=["Startjahr", "Verwendbare Erzeugung"]
    ).sort_values("Startjahr")

    if subset.empty:
        continue

    ax.plot(
        subset["Startjahr"].astype(float) + 0.5,
        subset["Verwendbare Erzeugung"],
        color=style["color"],
        marker=style["marker"],
        markersize=4,
        linewidth=1.8,
        label=style["label"],
        zorder=3,
    )


# Legends
crop_legend = ax.legend(
    title="Crop",
    loc="upper left",
    frameon=True,
)
ax.add_artist(crop_legend)

event_handles = [
    mpatches.Patch(
        facecolor=event_color,
        edgecolor="none",
        alpha=0.3,
        label="Event analysis window",
    ),
    plt.Line2D(
        [0], [0],
        color="#555555",
        linestyle=":",
        linewidth=1.2,
        label="Documented event start date",
    ),
]

ax.legend(
    handles=event_handles,
    title="Historical events",
    loc="upper right",
    fontsize=8,
    frameon=True,
)

ax.set_xticks(tick_years + 0.5)
ax.set_xticklabels(
    marketing_year_labels,
    rotation=60,
    ha="right",
    fontsize=8,
)

ax.set_xlim(first_year, plot_end_year)
ax.set_xlabel("Financial year (July–June)")
ax.set_ylabel("Usable production (unit as reported by BLE)")
ax.set_title(
    "Usable Production by Crop and Historical Event Windows",
    fontsize=14,
    pad=14,
)

ax.grid(axis="y", linestyle="--", alpha=0.3, zorder=1)

fig.text(
    0.01,
    0.01,
    "Shaded windows represent analytical event periods; dotted lines mark "
    "documented event starts where available. Temporal overlap does not "
    "demonstrate a causal effect on production.",
    fontsize=9,
    color="dimgray",
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()


# ============================================================
# FIGURE 2: PRE-CRISIS, CRISIS, AND POST-CRISIS PHASES
# ============================================================

fig, ax = plt.subplots(figsize=(17, 9))

phase_styles = {
    "Pre-crisis": {"color": "#007C83"},
    "Crisis": {"color": "#E87524"},
    "Post-crisis": {"color": "#7651A8"},
}

phase_order = ["Pre-crisis", "Crisis", "Post-crisis"]

phase_handles = [
    mpatches.Patch(
        facecolor=phase_styles[phase]["color"],
        edgecolor="none",
        label=phase,
    )
    for phase in phase_order
]

# Only events with usable intervals inside the BLE sample are displayed.
visible_events = [
    event for event in prepared_events
    if event["_clipped"] is not None
]

y_positions = np.arange(len(visible_events))

for y, event in zip(y_positions, visible_events):
    clipped_start, clipped_end = event["_clipped"]

    # The analysis window defines the crisis phase.
    crisis_start = clipped_start
    crisis_end = clipped_end

    # Pre-crisis phase: three years before the analytical start.
    pre_start = max(
        event["_start"] - pd.DateOffset(years=3),
        data_start,
    )
    pre_end = min(event["_start"], data_end_exclusive)

    pre_interval = clip_interval(pre_start, pre_end)

    if pre_interval is not None:
        start, end = pre_interval
        ax.barh(
            y,
            width=financial_year_x(end) - financial_year_x(start),
            left=financial_year_x(start),
            height=0.65,
            color=phase_styles["Pre-crisis"]["color"],
            edgecolor="none",
        )

    # Crisis / analysis phase
    ax.barh(
        y,
        width=financial_year_x(crisis_end) - financial_year_x(crisis_start),
        left=financial_year_x(crisis_start),
        height=0.65,
        color=phase_styles["Crisis"]["color"],
        edgecolor="none",
    )

    # Post-crisis: only when the event has a defined end date
    # and the event is not ongoing.
    if (
        event["_actual_end"] is not None
        or event.get("analysis_end_date") is not None
    ):
        post_start = event["_end"]
        post_end = min(
            event["_end"] + pd.DateOffset(years=3),
            data_end_exclusive,
        )

        post_interval = clip_interval(post_start, post_end)

        if post_interval is not None:
            start, end = post_interval
            ax.barh(
                y,
                width=financial_year_x(end) - financial_year_x(start),
                left=financial_year_x(start),
                height=0.65,
                color=phase_styles["Post-crisis"]["color"],
                edgecolor="none",
            )


# Event labels
event_labels = [
    f"{event.get('id', '')}: {event.get('name', 'Unnamed event')}"
    for event in visible_events
]

ax.set_yticks(y_positions)
ax.set_yticklabels(event_labels, fontsize=8)
ax.invert_yaxis()

ax.set_xticks(tick_years + 0.5)
ax.set_xticklabels(
    marketing_year_labels,
    rotation=60,
    ha="right",
    fontsize=8,
)

ax.set_xlim(first_year, plot_end_year)
ax.set_title(
    "Historical Events: Analytical Pre-crisis, Crisis, and Post-crisis Windows",
    fontsize=14,
    pad=14,
)
ax.set_xlabel("Financial year (July–June)")
ax.set_ylabel("Historical event")

ax.legend(
    handles=phase_handles,
    title="Analytical phases",
    loc="lower right",
    frameon=True,
)

ax.grid(axis="x", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

fig.text(
    0.01,
    0.01,
    "Pre-crisis and post-crisis windows are analytical comparison periods, "
    "not evidence of event impacts. Events outside the observed BLE period "
    "are omitted from this sample-aligned figure.",
    fontsize=9,
    color="dimgray",
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()
```
