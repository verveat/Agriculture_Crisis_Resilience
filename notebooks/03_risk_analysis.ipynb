```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ============================================================
# 1. TIME RANGE AND DATE HELPERS
# ============================================================

first_year = int(production_plot["Startjahr"].min())
last_year = int(production_plot["Startjahr"].max())
plot_end_year = last_year + 1

tick_years = np.arange(first_year, last_year + 1)
marketing_year_labels = [
    f"{year}/{str(year + 1)[-2:]}" for year in tick_years
]

data_start = pd.Timestamp(f"{first_year}-07-01")
data_end_exclusive = pd.Timestamp(f"{last_year + 1}-07-01")
data_end_inclusive = data_end_exclusive - pd.Timedelta(days=1)


def to_timestamp(value):
    """Convert a date-like value to Timestamp; return None if absent."""
    if value is None:
        return None

    try:
        if pd.isna(value) or str(value).strip() == "":
            return None
    except (TypeError, ValueError):
        pass

    return pd.Timestamp(value)


def financial_year_x(date):
    """
    Map a calendar date onto the continuous financial-year axis.
    July 1 = x.0; the following July 1 = x+1.0.
    """
    date = pd.Timestamp(date)
    fy_start_year = date.year if date.month >= 7 else date.year - 1
    fy_start = pd.Timestamp(f"{fy_start_year}-07-01")
    fy_end = pd.Timestamp(f"{fy_start_year + 1}-07-01")

    return fy_start_year + (
        (date - fy_start).total_seconds()
        / (fy_end - fy_start).total_seconds()
    )


def clip_interval(start, end):
    """Clip an interval to the observed BLE sample."""
    start = max(start, data_start)
    end = min(end, data_end_exclusive)

    if start >= end:
        return None

    return start, end


# ============================================================
# 2. PRODUCTION LINE STYLES
# ============================================================

product_styles = {
    "Weichweizen": {
        "color": "steelblue",
        "marker": "o",
        "label": "Soft wheat",
    },
    "Gerste": {
        "color": "darkorange",
        "marker": "s",
        "label": "Barley",
    },
    "Kartoffeln": {
        "color": "forestgreen",
        "marker": "^",
        "label": "Potatoes",
    },
}

event_color = "#8C8C8C"


# ============================================================
# 3. PREPARE EVENT WINDOWS
# ============================================================

prepared_events = []

for event in events:
    analysis_start = to_timestamp(event.get("analysis_start_date"))
    analysis_end = to_timestamp(event.get("analysis_end_date"))

    actual_start = to_timestamp(event.get("event_start_date"))
    actual_end = to_timestamp(event.get("event_end_date"))

    # Prefer analysis boundaries; fall back to actual event dates.
    start = (
        analysis_start
        if analysis_start is not None
        else actual_start
    )
    end = (
        analysis_end
        if analysis_end is not None
        else actual_end
    )

    if start is None:
        print(f"Skipping {event.get('id')}: no start date available.")
        continue

    # Exclude events starting at or after the sample boundary.
    if start >= data_end_exclusive:
        print(
            f"Excluding {event.get('id')}: "
            "outside production data coverage."
        )
        continue

    # Cap open-ended or later analysis windows at the sample boundary.
    if end is None or end > data_end_exclusive:
        end = data_end_exclusive

    clipped = clip_interval(start, end)

    if clipped is None:
        print(f"Skipping {event.get('id')}: no overlap with data.")
        continue

    prepared_events.append({
        **event,
        "_start": start,
        "_end": end,
        "_actual_start": actual_start,
        "_actual_end": actual_end,
        "_clipped": clipped,
    })

prepared_events.sort(key=lambda e: e["_start"])

print(f"Observed production period: {first_year}/{str(first_year + 1)[-2:]} "
      f"to {last_year}/{str(last_year + 1)[-2:]}")
print(f"Events included in the plots: {len(prepared_events)}")
print(f"Data coverage ends: {data_end_inclusive.date()}")


# ============================================================
# FIGURE 1: PRODUCTION AND EVENT ANALYSIS WINDOWS
# ============================================================

fig, ax = plt.subplots(figsize=(17, 8))

# Analytical windows are shaded only within the observed data period.
for event in prepared_events:
    interval = event["_clipped"]

    if interval is None:
        continue

    start, end = interval
    x_start = financial_year_x(start)
    x_end = financial_year_x(end)

    ax.axvspan(
        x_start,
        x_end,
        color=event_color,
        alpha=0.13,
        linewidth=0,
        zorder=0,
    )

    # Mark documented actual event starts within the observed sample.
    actual_start = event["_actual_start"]

    if (
        actual_start is not None
        and data_start <= actual_start < data_end_exclusive
    ):
        ax.axvline(
            financial_year_x(actual_start),
            color="#555555",
            linewidth=0.8,
            linestyle=":",
            alpha=0.75,
            zorder=1,
        )


# Production lines
for product, style in product_styles.items():
    subset = production_plot[
        production_plot["Bilanz Produkt"] == product
    ].copy()

    subset = subset.dropna(
        subset=["Startjahr", "Verwendbare Erzeugung"]
    ).sort_values("Startjahr")

    if subset.empty:
        print(f"Warning: no production data found for {product}.")
        continue

    ax.plot(
        subset["Startjahr"].astype(float) + 0.5,
        subset["Verwendbare Erzeugung"],
        color=style["color"],
        marker=style["marker"],
        markersize=4,
        linewidth=1.8,
        label=style["label"],
        zorder=3,
    )


# Crop legend
crop_legend = ax.legend(
    title="Crop",
    loc="upper left",
    frameon=True,
)
ax.add_artist(crop_legend)

# Event legend
event_handles = [
    mpatches.Patch(
        facecolor=event_color,
        edgecolor="none",
        alpha=0.3,
        label="Event analysis window",
    ),
    plt.Line2D(
        [0],
        [0],
        color="#555555",
        linestyle=":",
        linewidth=1.2,
        label="Documented event start date",
    ),
]

ax.legend(
    handles=event_handles,
    title="Historical events",
    loc="upper right",
    fontsize=8,
    frameon=True,
)

ax.set_xticks(tick_years + 0.5)
ax.set_xticklabels(
    marketing_year_labels,
    rotation=60,
    ha="right",
    fontsize=8,
)

ax.set_xlim(first_year, plot_end_year)
ax.set_xlabel("Financial year (July–June)")
ax.set_ylabel("Usable production (unit as reported by BLE)")
ax.set_title(
    "Usable Production by Crop and Historical Event Windows",
    fontsize=14,
    pad=14,
)

ax.grid(axis="y", linestyle="--", alpha=0.3, zorder=1)

fig.text(
    0.01,
    0.01,
    "Shaded areas represent analytical event windows; dotted lines mark "
    "documented event start dates. Temporal overlap does not demonstrate "
    "a causal effect on production.",
    fontsize=9,
    color="dimgray",
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()


# ============================================================
# FIGURE 2: ANALYTICAL PHASES
# ============================================================

fig, ax = plt.subplots(figsize=(17, 9))

phase_styles = {
    "Pre-analysis": {"color": "#007C83"},
    "Analysis window": {"color": "#E87524"},
    "Post-analysis": {"color": "#7651A8"},
}

phase_order = [
    "Pre-analysis",
    "Analysis window",
    "Post-analysis",
]

phase_handles = [
    mpatches.Patch(
        facecolor=phase_styles[phase]["color"],
        edgecolor="none",
        label=phase,
    )
    for phase in phase_order
]

# All displayed events overlap the observed BLE sample.
visible_events = [
    event for event in prepared_events
    if event["_clipped"] is not None
]

y_positions = np.arange(len(visible_events))

for y, event in zip(y_positions, visible_events):
    clipped_start, clipped_end = event["_clipped"]

    # Analysis window
    ax.barh(
        y,
        width=(
            financial_year_x(clipped_end)
            - financial_year_x(clipped_start)
        ),
        left=financial_year_x(clipped_start),
        height=0.65,
        color=phase_styles["Analysis window"]["color"],
        edgecolor="none",
    )

    # Pre-analysis: up to three years before the analysis window.
    pre_start = max(
        event["_start"] - pd.DateOffset(years=3),
        data_start,
    )
    pre_end = min(event["_start"], data_end_exclusive)

    pre_interval = clip_interval(pre_start, pre_end)

    if pre_interval is not None:
        start, end = pre_interval

        ax.barh(
            y,
            width=(
                financial_year_x(end)
                - financial_year_x(start)
            ),
            left=financial_year_x(start),
            height=0.65,
            color=phase_styles["Pre-analysis"]["color"],
            edgecolor="none",
        )

    # Post-analysis: only for explicitly bounded analysis windows.
    # Open-ended windows are not assigned an artificial post-period.
    has_defined_analysis_end = (
        to_timestamp(event.get("analysis_end_date")) is not None
    )

    if has_defined_analysis_end:
        post_start = event["_end"]
        post_end = min(
            event["_end"] + pd.DateOffset(years=3),
            data_end_exclusive,
        )

        post_interval = clip_interval(post_start, post_end)

        if post_interval is not None:
            start, end = post_interval

            ax.barh(
                y,
                width=(
                    financial_year_x(end)
                    - financial_year_x(start)
                ),
                left=financial_year_x(start),
                height=0.65,
                color=phase_styles["Post-analysis"]["color"],
                edgecolor="none",
            )


# Event labels
event_labels = [
    f"{event.get('id', '')}: {event.get('name', 'Unnamed event')}"
    for event in visible_events
]

ax.set_yticks(y_positions)
ax.set_yticklabels(event_labels, fontsize=8)
ax.invert_yaxis()

ax.set_xticks(tick_years + 0.5)
ax.set_xticklabels(
    marketing_year_labels,
    rotation=60,
    ha="right",
    fontsize=8,
)

ax.set_xlim(first_year, plot_end_year)

ax.set_title(
    "Historical Events: Analytical Comparison Phases",
    fontsize=14,
    pad=14,
)

ax.set_xlabel("Financial year (July–June)")
ax.set_ylabel("Historical event")

# Legend in the lower-left corner, as requested.
ax.legend(
    handles=phase_handles,
    title="Analytical phases",
    loc="lower left",
    frameon=True,
)

ax.grid(axis="x", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

fig.text(
    0.01,
    0.01,
    "Pre-analysis and post-analysis periods are comparison windows, "
    "not evidence of event impacts. Open-ended analysis windows are "
    "capped at the observed data boundary. Events starting after the "
    "observed sample are excluded.",
    fontsize=9,
    color="dimgray",
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()
```

```python
import pandas as pd
from pathlib import Path

project_dir = Path.cwd().parent
file_path = (
    project_dir
    / "data"
    / "processed"
    / "agricultural_risk_event_database_1991-2025.xlsx"
)

events_review = pd.read_excel(file_path, sheet_name="events")

date_columns = [
    "event_start_date",
    "event_end_date",
    "analysis_start_date",
    "analysis_end_date",
]

for col in date_columns:
    events_review[col] = pd.to_datetime(
        events_review[col], errors="coerce"
    )

# Beobachteter Produktionszeitraum
data_start = pd.Timestamp("1991-07-01")
data_end = pd.Timestamp("2025-06-30")

def assess_window(row):
    start = row["analysis_start_date"]
    end = row["analysis_end_date"]

    if pd.isna(start):
        return "Prüfen: Analysebeginn fehlt"

    if start > data_end:
        return "Außerhalb der aktuellen Datenabdeckung"

    if pd.notna(end) and end < data_start:
        return "Außerhalb der aktuellen Datenabdeckung"

    if pd.isna(end):
        return "Offenes Fenster; nur bis Datenende auswerten"

    if end > data_end:
        return "Fenster teilweise außerhalb der Datenabdeckung"

    if start < data_start:
        return "Fenster beginnt vor der Datenabdeckung"

    return "Innerhalb der Datenabdeckung"

events_review["coverage_check"] = events_review.apply(
    assess_window, axis=1
)

display(
    events_review[
        [
            "id",
            "name",
            "analysis_start_date",
            "analysis_end_date",
            "event_start_date",
            "event_end_date",
            "coverage_check",
        ]
    ].sort_values("id")
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path

# ============================================================
# 0. CONFIGURATION AND INPUT CHECKS
# ============================================================

project_dir = Path.cwd().parent
processed_dir = project_dir / "data" / "processed"

database_path = (
    processed_dir
    / "agricultural_risk_event_database_1991-2025.xlsx"
)

if "production_plot" not in globals():
    raise NameError(
        "production_plot fehlt. Bitte zuerst die Zelle ausführen, "
        "die die BLE-Produktionsdaten vorbereitet."
    )

production_plot = production_plot.copy()

# Use prepared events if available; otherwise read the Excel file.
if "events" in globals() and isinstance(events, list):
    events_input = [dict(event) for event in events]
else:
    if not database_path.exists():
        raise FileNotFoundError(
            f"Ereignisdatenbank nicht gefunden: {database_path}\n"
            "Bitte den Dateinamen und den Ordner data/processed prüfen."
        )

    events_input = pd.read_excel(
        database_path,
        sheet_name="events"
    ).to_dict("records")

if production_plot.empty:
    raise ValueError("production_plot enthält keine Produktionsdaten.")

required_columns = {
    "Startjahr",
    "Bilanz Produkt",
    "Verwendbare Erzeugung",
}

missing_columns = required_columns - set(production_plot.columns)

if missing_columns:
    raise ValueError(
        f"Fehlende Spalten in production_plot: {missing_columns}"
    )

if not events_input:
    raise ValueError("Die Ereignisdatenbank enthält keine Ereignisse.")

# ============================================================
# 1. PREPARE PRODUCTION DATA
# ============================================================

production_plot["Startjahr"] = pd.to_numeric(
    production_plot["Startjahr"],
    errors="coerce"
)

production_plot["Verwendbare Erzeugung"] = pd.to_numeric(
    production_plot["Verwendbare Erzeugung"],
    errors="coerce"
)

production = production_plot.dropna(
    subset=[
        "Startjahr",
        "Verwendbare Erzeugung",
        "Bilanz Produkt",
    ]
).copy()

production["Startjahr"] = production["Startjahr"].astype(int)

first_year = int(production["Startjahr"].min())
last_year = int(production["Startjahr"].max())
plot_end_year = last_year + 1

# Observed sample boundaries:
# 1 July of the first year through 30 June of the final year.
data_start = pd.Timestamp(f"{first_year}-07-01")
data_end_exclusive = pd.Timestamp(f"{last_year + 1}-07-01")
data_end_inclusive = data_end_exclusive - pd.Timedelta(days=1)

tick_years = np.arange(first_year, last_year + 1, 2)

if last_year not in tick_years:
    tick_years = np.append(tick_years, last_year)

marketing_year_labels = [
    f"{year}/{str(year + 1)[-2:]}"
    for year in tick_years
]

product_styles = {
    "Weichweizen": {
        "color": "steelblue",
        "marker": "o",
        "label": "Soft wheat",
    },
    "Gerste": {
        "color": "darkorange",
        "marker": "s",
        "label": "Barley",
    },
    "Kartoffeln": {
        "color": "forestgreen",
        "marker": "^",
        "label": "Potatoes",
    },
}

# ============================================================
# 2. DATE HELPERS
# ============================================================

def to_timestamp(value):
    """Convert a date value to Timestamp; missing/invalid values -> NaT."""
    if value is None:
        return pd.NaT

    try:
        if pd.isna(value) or str(value).strip() == "":
            return pd.NaT

        return pd.Timestamp(value)

    except (TypeError, ValueError):
        return pd.NaT


def financial_year_x(date):
    """
    Convert a date to the continuous financial-year axis.

    1 July = x.0
    1 January = approximately x.5
    Following 1 July = x+1.0
    """
    date = pd.Timestamp(date)

    fy_start_year = (
        date.year if date.month >= 7 else date.year - 1
    )

    fy_start = pd.Timestamp(f"{fy_start_year}-07-01")
    fy_end = pd.Timestamp(f"{fy_start_year + 1}-07-01")

    fraction = (
        (date - fy_start).total_seconds()
        / (fy_end - fy_start).total_seconds()
    )

    return fy_start_year + fraction


def inclusive_end_to_exclusive(end):
    """
    Convert an inclusive end date to an exclusive boundary.
    For example, 30 June becomes 1 July.
    """
    if pd.isna(end):
        return pd.NaT

    return (
        pd.Timestamp(end).normalize()
        + pd.Timedelta(days=1)
    )


def clip_interval(start, end):
    """Restrict a half-open interval to the observed data period."""
    if pd.isna(start) or pd.isna(end):
        return None

    start = max(pd.Timestamp(start), data_start)
    end = min(pd.Timestamp(end), data_end_exclusive)

    if start >= end:
        return None

    return start, end


# ============================================================
# 3. PREPARE EVENT WINDOWS
# ============================================================

prepared_events = []

for event in events_input:

    analysis_start = to_timestamp(
        event.get("analysis_start_date")
    )
    analysis_end = to_timestamp(
        event.get("analysis_end_date")
    )

    actual_start = to_timestamp(
        event.get("event_start_date")
    )
    actual_end = to_timestamp(
        event.get("event_end_date")
    )

    # Prefer the analytical window for plotting.
    # Use actual dates only when analytical boundaries are absent.
    start = (
        analysis_start
        if pd.notna(analysis_start)
        else actual_start
    )

    raw_end = (
        analysis_end
        if pd.notna(analysis_end)
        else actual_end
    )

    if pd.isna(start):
        print(
            f"Skipping {event.get('id', '?')}: "
            "no usable start date."
        )
        continue

    # Exclude events that begin on or after the end of the sample.
    if start >= data_end_exclusive:
        print(
            f"Excluding {event.get('id', '?')} from the plots: "
            "outside the observed production period."
        )
        continue

    # Treat supplied end dates as inclusive.
    # Open-ended windows are capped at the sample boundary.
    if pd.notna(raw_end):
        end = inclusive_end_to_exclusive(raw_end)
    else:
        end = data_end_exclusive

    clipped = clip_interval(start, end)

    if clipped is None:
        print(
            f"Skipping {event.get('id', '?')}: "
            "no overlap with the observed production period."
        )
        continue

    prepared_events.append({
        **event,
        "_start": start,
        "_end": end,
        "_actual_start": actual_start,
        "_actual_end": actual_end,
        "_analysis_start": analysis_start,
        "_analysis_end": analysis_end,
        "_clipped": clipped,
    })

prepared_events.sort(key=lambda event: event["_start"])

visible_events = [
    event
    for event in prepared_events
    if event["_clipped"] is not None
]

if not visible_events:
    raise ValueError(
        "Kein Ereignis überschneidet sich mit dem beobachteten "
        "Produktionszeitraum."
    )

# ============================================================
# 4. FIGURE 1
# USABLE PRODUCTION AND HISTORICAL EVENT WINDOWS
# ============================================================

fig, ax = plt.subplots(figsize=(17, 8))

event_color = "#8C8C8C"

# 4a. Analytical windows
for event in visible_events:
    start, end = event["_clipped"]

    ax.axvspan(
        financial_year_x(start),
        financial_year_x(end),
        color=event_color,
        alpha=0.13,
        linewidth=0,
        zorder=0,
    )

# 4b. Production series
for product, style in product_styles.items():

    subset = production[
        production["Bilanz Produkt"] == product
    ].sort_values("Startjahr")

    if subset.empty:
        continue

    ax.plot(
        subset["Startjahr"].astype(float) + 0.5,
        subset["Verwendbare Erzeugung"],
        color=style["color"],
        marker=style["marker"],
        markersize=4,
        linewidth=1.8,
        label=style["label"],
        zorder=3,
    )

# 4c. Documented actual event start dates
for event in visible_events:
    actual_start = event["_actual_start"]

    if (
        pd.notna(actual_start)
        and data_start <= actual_start < data_end_exclusive
    ):
        ax.axvline(
            financial_year_x(actual_start),
            color="#555555",
            linewidth=0.9,
            linestyle=":",
            alpha=0.75,
            zorder=2,
        )

# 4d. Separate legends for crops and historical events
crop_handles, crop_labels = ax.get_legend_handles_labels()

crop_legend = ax.legend(
    crop_handles,
    crop_labels,
    title="Crop",
    loc="upper left",
    frameon=True,
)

ax.add_artist(crop_legend)

event_handles = [
    mpatches.Patch(
        facecolor=event_color,
        edgecolor="none",
        alpha=0.3,
        label="Event analysis window",
    ),
    plt.Line2D(
        [0], [0],
        color="#555555",
        linestyle=":",
        linewidth=1.2,
        label="Documented event start date",
    ),
]

ax.legend(
    handles=event_handles,
    title="Historical events",
    loc="upper right",
    fontsize=8,
    frameon=True,
)

# 4e. Financial-year axis and labels
ax.set_xticks(tick_years + 0.5)
ax.set_xticklabels(
    marketing_year_labels,
    rotation=55,
    ha="right",
    fontsize=8,
)

ax.set_xlim(first_year, plot_end_year)

ax.set_xlabel("Financial year (1 July–30 June)")
ax.set_ylabel(
    "Usable production (unit as reported by BLE; verify metadata)"
)
ax.set_title(
    "Usable Production by Crop and Historical Event Windows",
    fontsize=14,
    pad=14,
)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.3,
    zorder=1,
)

fig.text(
    0.01,
    0.01,
    "Shaded areas indicate analytical windows, not necessarily actual "
    "event durations. Dotted lines indicate documented event start dates. "
    "Temporal overlap does not establish causality.",
    fontsize=9,
    color="dimgray",
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()


# ============================================================
# 5. FIGURE 2
# PRE-ANALYSIS, ANALYSIS, AND POST-ANALYSIS WINDOWS
# ============================================================

fig, ax = plt.subplots(figsize=(17, 9))

phase_styles = {
    "Pre-analysis": {"color": "#007C83"},
    "Analysis window": {"color": "#E87524"},
    "Post-analysis": {"color": "#7651A8"},
}

phase_order = [
    "Pre-analysis",
    "Analysis window",
    "Post-analysis",
]

phase_handles = [
    mpatches.Patch(
        facecolor=phase_styles[phase]["color"],
        edgecolor="none",
        label=phase,
    )
    for phase in phase_order
]

y_positions = np.arange(len(visible_events))

for y, event in zip(y_positions, visible_events):

    clipped_start, clipped_end = event["_clipped"]

    # Pre-analysis comparison period: up to three years before
    # the analytical start, clipped to the observed sample.
    pre_start = max(
        event["_start"] - pd.DateOffset(years=3),
        data_start,
    )
    pre_end = min(
        event["_start"],
        data_end_exclusive,
    )

    pre_interval = clip_interval(pre_start, pre_end)

    if pre_interval is not None:
        start, end = pre_interval

        ax.barh(
            y,
            width=(
                financial_year_x(end)
                - financial_year_x(start)
            ),
            left=financial_year_x(start),
            height=0.65,
            color=phase_styles["Pre-analysis"]["color"],
            edgecolor="none",
        )

    # Main analytical window
    ax.barh(
        y,
        width=(
            financial_year_x(clipped_end)
            - financial_year_x(clipped_start)
        ),
        left=financial_year_x(clipped_start),
        height=0.65,
        color=phase_styles["Analysis window"]["color"],
        edgecolor="none",
    )

    # Post-analysis comparison period: only if an analytical end
    # date was explicitly recorded. This does not imply that the
    # historical event itself ended at that date.
    if pd.notna(event["_analysis_end"]):

        post_start = event["_end"]
        post_end = min(
            post_start + pd.DateOffset(years=3),
            data_end_exclusive,
        )

        post_interval = clip_interval(post_start, post_end)

        if post_interval is not None:
            start, end = post_interval

            ax.barh(
                y,
                width=(
                    financial_year_x(end)
                    - financial_year_x(start)
                ),
                left=financial_year_x(start),
                height=0.65,
                color=phase_styles["Post-analysis"]["color"],
                edgecolor="none",
            )

# Event labels
event_labels = [
    f"{event.get('id', '')}: {event.get('name', 'Unnamed event')}"
    for event in visible_events
]

ax.set_yticks(y_positions)
ax.set_yticklabels(event_labels, fontsize=8)
ax.invert_yaxis()

ax.set_xticks(tick_years + 0.5)
ax.set_xticklabels(
    marketing_year_labels,
    rotation=55,
    ha="right",
    fontsize=8,
)

ax.set_xlim(first_year, plot_end_year)

ax.set_title(
    "Historical Events and Analytical Comparison Windows",
    fontsize=14,
    pad=14,
)
ax.set_xlabel("Financial year (1 July–30 June)")
ax.set_ylabel("Historical event")

ax.legend(
    handles=phase_handles,
    title="Analytical phases",
    loc="lower left",
    frameon=True,
)

ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.3,
)
ax.set_axisbelow(True)

fig.text(
    0.01,
    0.01,
    "Pre-analysis and post-analysis intervals are comparison periods, "
    "not evidence of impacts or actual event duration. Open-ended "
    "analytical windows have no post-analysis interval.",
    fontsize=9,
    color="dimgray",
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()


# ============================================================
# 6. DIAGNOSTICS: PRODUCTION DATA
# ============================================================

print("=" * 65)
print("OBSERVED PRODUCTION SAMPLE")
print("=" * 65)

print(f"First financial year: {first_year}/{str(first_year + 1)[-2:]}")
print(f"Last financial year:  {last_year}/{str(last_year + 1)[-2:]}")
print(f"Number of financial years: {last_year - first_year + 1}")
print(f"Number of products: {production['Bilanz Produkt'].nunique()}")
print(f"Missing production values after cleaning: "
      f"{production['Verwendbare Erzeugung'].isna().sum()}")

production_diagnostics = (
    production.groupby("Bilanz Produkt")["Verwendbare Erzeugung"]
    .agg(
        observations="count",
        minimum="min",
        median="median",
        maximum="max",
    )
    .reset_index()
)

display(production_diagnostics)

print(
    "\nUNIT CHECK: The code does not infer or convert the BLE unit. "
    "Verify the source metadata and confirm whether the units are "
    "comparable across products before interpreting absolute levels."
)


# ============================================================
# 7. DIAGNOSTICS: EVENT DATES AND SOURCES
# ============================================================

def source_details(value):
    """Format source information consistently for the audit table."""
    if value is None:
        return ""

    if isinstance(value, (list, tuple)):
        return "; ".join(str(item) for item in value)

    try:
        if pd.isna(value):
            return ""
    except (TypeError, ValueError):
        pass

    text = str(value).strip()

    if text.lower() in ("", "nan", "none", "[]"):
        return ""

    return text


event_audit_rows = []

for event in prepared_events:

    actual_start = event["_actual_start"]
    actual_end = event["_actual_end"]
    source_text = source_details(event.get("sources"))

    event_audit_rows.append({
        "id": event.get("id"),
        "name": event.get("name"),
        "actual_start_date": actual_start,
        "actual_end_date": actual_end,
        "start_date_needs_review": pd.isna(actual_start),
        "end_date_needs_review": pd.isna(actual_end),
        "source_details": source_text,
        "source_recorded": bool(source_text),
        "shown_in_current_sample": event["_clipped"] is not None,
    })

event_audit = pd.DataFrame(event_audit_rows)

print("\n" + "=" * 65)
print("EVENT DATE AND SOURCE AUDIT")
print("=" * 65)

display(
    event_audit[
        [
            "id",
            "name",
            "actual_start_date",
            "actual_end_date",
            "start_date_needs_review",
            "end_date_needs_review",
            "source_recorded",
            "shown_in_current_sample",
        ]
    ]
)

print(
    "\nInterpretation note: A missing actual event date is not "
    "automatically an error. For ongoing or diffuse developments, "
    "retain the uncertainty rather than inventing a date."
)

print("\nEvents included in the plots:", len(visible_events))
print("Observed data end date:", data_end_inclusive.date())
```
