```python

# Import libraries
from pathlib import Path

import pandas as pd
```

```python

# Define project directory
project_dir = Path.cwd().parent

# Define source file path
file_path = (
    project_dir
    / "data"
    / "raw"
    / "bmelh_ble_versorgungsbilanzen_1990-2025_weichweizen_gerste_kartoffel"
    / "ble_versorgungsbilanzen_1991-2025"
    / "01_ready_to_clean_ble_supply_balance_all_products_1991-2025.xls"
)

# Check whether the source file exists
print(f"Project directory: {project_dir}")
print(f"Source file exists: {file_path.exists()}")
print(f"Source file path: {file_path}")
```

```python

# Load source data
df = pd.read_excel(file_path, engine="xlrd")

# Display the first five rows
display(df.head())
```

```python

# Inspect dataset dimensions
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")

# Inspect column names
print("\nColumn names:")
print(df.columns.tolist())

# Inspect data types and non-null values
print("\nDataset information:")
df.info()
```

```python

# Display the first ten rows
display(df.head(10))
```

```python

# Check missing values
print("Missing values per column:")
display(df.isna().sum().to_frame(name="Missing values"))

# Check duplicate rows
duplicate_count = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicate_count}")

# Check unique values per column
print("\nNumber of unique values per column:")
display(df.nunique().to_frame(name="Unique values"))
```

```python

# Inspect rows with missing balance values
missing_values = (
    df.loc[
        df["Bilanz Wert"].isna(),
        [
            "Wirtschaftsjahr",
            "Bilanz Produkt",
            "Bilanz Oberbegriff",
            "Bilanz Position",
            "Bilanz Einheit",
        ],
    ]
    .sort_values(
        ["Bilanz Produkt", "Bilanz Position", "Wirtschaftsjahr"]
    )
)

# Display all rows with missing values
display(missing_values)
```

```python

# Summarize missing values by product and balance position
missing_summary = (
    df.loc[df["Bilanz Wert"].isna()]
    .groupby(
        [
            "Bilanz Produkt",
            "Bilanz Oberbegriff",
            "Bilanz Position",
            "Bilanz Einheit",
        ],
        dropna=False,
    )
    .size()
    .reset_index(name="Missing values")
    .sort_values(
        ["Bilanz Produkt", "Bilanz Position"]
    )
)

display(missing_summary)
```

```python

# Count missing values by product, balance position, and year
missing_by_year = (
    df.loc[df["Bilanz Wert"].isna()]
    .groupby(
        [
            "Bilanz Produkt",
            "Bilanz Position",
            "Wirtschaftsjahr",
        ],
        dropna=False,
    )
    .size()
    .reset_index(name="Missing values")
    .sort_values(
        [
            "Bilanz Produkt",
            "Bilanz Position",
            "Wirtschaftsjahr",
        ]
    )
)

display(missing_by_year)
```

```python

# Summarize missing year ranges by product and balance position
missing_year_ranges = (
    df.loc[df["Bilanz Wert"].isna()]
    .groupby(
        ["Bilanz Produkt", "Bilanz Position"],
        dropna=False,
    )
    .agg(
        first_missing_year=("Wirtschaftsjahr", "min"),
        last_missing_year=("Wirtschaftsjahr", "max"),
        number_of_missing_years=("Wirtschaftsjahr", "nunique"),
    )
    .reset_index()
)

display(missing_year_ranges)
```

```python

# Count records by product
product_counts = (
    df.groupby("Bilanz Produkt")
    .size()
    .reset_index(name="Number of records")
    .sort_values("Bilanz Produkt")
)

display(product_counts)

# Check whether all products have the same number of records
same_record_count = product_counts["Number of records"].nunique() == 1

print(f"All products have the same number of records: {same_record_count}")
```

```python
# Data quality note: The combined dataset contains 1,646 records: 578 for barley, 490 for potatoes, and 578 for soft wheat. The differences in record counts were checked against the original source data and were confirmed to exist in the raw data. Therefore, the differences were not caused by the data consolidation process. Missing values are retained for further investigation rather than being automatically replaced or removed.
```

```python
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Check year coverage by product
year_coverage = (
    df.groupby("Bilanz Produkt")
    .agg(
        Number_of_records=("Wirtschaftsjahr", "size"),
        Unique_years=("Wirtschaftsjahr", "nunique"),
        First_year=("Wirtschaftsjahr", "min"),
        Last_year=("Wirtschaftsjahr", "max"),
    )
    .reset_index()
)

display(year_coverage)
```

```python
# Remove financial year 1990/91
df = df[df["Wirtschaftsjahr"] != "1990/91"].copy()
```

```python
# Check year coverage by product
year_coverage = (
    df.groupby("Bilanz Produkt")
    .agg(
        Number_of_records=("Wirtschaftsjahr", "size"),
        Unique_years=("Wirtschaftsjahr", "nunique"),
        First_year=("Wirtschaftsjahr", "min"),
        Last_year=("Wirtschaftsjahr", "max"),
    )
    .reset_index()
)

display(year_coverage)
```

```python
# Check the dataset dimensions
print(df.shape)
```

```python
# Count records per product and financial year
records_per_year = (
    df.groupby(["Wirtschaftsjahr", "Bilanz Produkt"])
    .size()
    .unstack(fill_value=0)
    .sort_index()
)

display(records_per_year)
```

```python

# Compare balance positions across products for a common year
comparison_year = "1991/92"

position_columns = [
    "Bilanz Oberbegriff",
    "Bilanz Position",
    "Bilanz Einheit",
]

year_df = df.loc[
    df["Wirtschaftsjahr"] == comparison_year,
    ["Bilanz Produkt"] + position_columns,
].copy()

# Normalize whitespace for comparison only
for column in position_columns:
    year_df[column] = (
        year_df[column]
        .astype("string")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

# Extract unique balance positions for each product
positions_by_product = {}

for product in ["Gerste", "Weichweizen", "Kartoffeln"]:
    positions = (
        year_df.loc[
            year_df["Bilanz Produkt"] == product,
            position_columns,
        ]
        .drop_duplicates()
        .sort_values(position_columns)
        .reset_index(drop=True)
    )

    positions_by_product[product] = positions

    print(f"\n{product}: {len(positions)} unique balance positions")
    display(positions)
```

```python
# Data quality note: The dataset contains 17 records per year for barley and soft wheat and 14 records per year for potatoes. A comparison of balance positions for the financial year 1991/92 showed that the product balances differ in their structure and include product-specific positions and differently named categories. The differences in record counts were also verified against the original source data. No records have been deleted, and no missing values have been replaced. The original data structure is retained for further analysis.
```

```python
# Count records by product
product_counts = df["Bilanz Produkt"].value_counts().sort_index()

display(product_counts)
```

```python
# Standardize whitespace in text columns
text_columns = df.select_dtypes(include=["str"]).columns

for column in text_columns:
    df[column] = (
        df[column]
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

# Check dataset dimensions
print(f"Dataset shape: {df.shape}")
```

```python
# Final data quality checks
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Missing values in Bilanz Wert: {df['Bilanz Wert'].isna().sum()}")

# Count records by product
display(df["Bilanz Produkt"].value_counts().sort_index())

# Inspect data types
display(df.dtypes)
```
