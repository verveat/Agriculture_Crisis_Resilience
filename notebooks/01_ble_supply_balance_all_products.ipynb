```python

import pandas as pd
import numpy as np

# 1. Define the required balance positions
required_positions = [
    "Einfuhr",
    "Ausfuhr",
    "Inlandsverwendung insgesamt",
    "Gesamterzeugung",
]

# 2. Select the required rows
selected = df[
    df["Bilanz Position"].isin(required_positions)
].copy()

# 3. Reshape the data by year and product
quota_data = selected.pivot_table(
    index=["Wirtschaftsjahr", "Bilanz Produkt"],
    columns="Bilanz Position",
    values="Bilanz Wert",
    aggfunc="first"
)

# 4. Calculate the ratios
quota_data["Importquote"] = (
    quota_data["Einfuhr"]
    / quota_data["Inlandsverwendung insgesamt"]
    * 100
)

quota_data["Exportquote"] = (
    quota_data["Ausfuhr"]
    / quota_data["Gesamterzeugung"]
    * 100
)

# 5. Exclude undefined or infinite results
quota_data[["Importquote", "Exportquote"]] = (
    quota_data[["Importquote", "Exportquote"]]
    .replace([np.inf, -np.inf], np.nan)
)

# 6. Convert the ratios into additional balance rows
new_rows = []

for (year, product), row in quota_data.iterrows():
    for position in ["Importquote", "Exportquote"]:
        value = row[position]

        if pd.notna(value):
            new_rows.append({
                "Wirtschaftsjahr": year,
                "Bilanz Produkt": product,
                "Bilanz Oberbegriff": "Berechnete Kennzahl",
                "Bilanz Position": position,
                "Bilanz Einheit": "%",
                "Bilanz Wert": value
            })

# 7. Append the new rows without modifying the original df
df_with_quotas = pd.concat(
    [df, pd.DataFrame(new_rows, columns=df.columns)],
    ignore_index=True
)

# 8. Display the calculated ratios
quota_results = df_with_quotas[
    df_with_quotas["Bilanz Position"].isin(
        ["Importquote", "Exportquote"]
    )
].copy()

display(
    quota_results.sort_values(
        ["Bilanz Produkt", "Wirtschaftsjahr", "Bilanz Position"]
    ).round({"Bilanz Wert": 2})
)

# 9. Verify the result
print(f"Original rows: {len(df)}")
print(f"Added quota rows: {len(new_rows)}")
print(f"Extended dataset rows: {len(df_with_quotas)}")
```

### Methodological Note: Bilanz Position Comparability

Before calculating import and export ratios, the availability of relevant balance positions was checked across barley, potatoes, and soft wheat.

| Balance Position | Barley | Potatoes | Soft Wheat |
|---|---|---|---|
| Imports | ✓ | ✓ | ✓ |
| Exports | ✓ | ✓ | ✓ |
| Domestic Use | Total | Available | Total |
| Total Production | — | ✓ | — |
| Usable Production | ✓ | ✓ | ✓ |
| Self-sufficiency Rate | ✓ | ✓ | ✓ |

**Methodological decision:** Usable production is available for all three products and will be assessed as a common denominator for the export ratio. For the import ratio, the domestic-use positions must first be confirmed as comparable using official BLE definitions. The original BLE self-sufficiency rates will be retained unchanged.

All calculated ratios will be stored separately to preserve the original dataset.

```python

# Select the required balance sheet items for the ratios
positions = [
    "Einfuhr",
    "Ausfuhr",
    "Inlandsverwendung",
    "Inlandsverwendung insgesamt",
    "Verwendbare Erzeugung",
]

quota_input = df.loc[
    df["Bilanz Position"].isin(positions),
    [
        "Wirtschaftsjahr",
        "Bilanz Produkt",
        "Bilanz Position",
        "Bilanz Wert",
    ],
].copy()

display(quota_input.head(10))

print("Zeilen:", len(quota_input))
print("Produkte:", quota_input["Bilanz Produkt"].unique())
```

### Methodology: Import and Export Ratios

Before calculating the ratios, the availability and uniqueness of the relevant balance positions were checked for all three products across 34 financial years. No duplicate combinations were found.

**Calculation rules**

- **Import Ratio (%)** = Imports / Domestic Use × 100
- **Export Ratio (%)** = Exports / Usable Production × 100

The domestic-use position differs by product: `Inlandsverwendung insgesamt` is used for barley and soft wheat, while `Inlandsverwendung` is used for potatoes.

`Verwendbare Erzeugung` is selected as the common denominator for the export ratio because it is available for all three products. `Gesamterzeugung`, available only for potatoes, is excluded to maintain consistency pending confirmation of the official definitions.

The ratios are self-calculated analytical indicators, not official BLE indicators. The original dataset and published self-sufficiency rates remain unchanged.

```python

# Define the relevant balance positions for each product
domestic_use_positions = {
    "Gerste": "Inlandsverwendung insgesamt",
    "Kartoffeln": "Inlandsverwendung",
    "Weichweizen": "Inlandsverwendung insgesamt",
}

# Select the required balance positions
required_positions = [
    "Einfuhr",
    "Ausfuhr",
    "Inlandsverwendung",
    "Inlandsverwendung insgesamt",
    "Verwendbare Erzeugung",
]

quota_input = df.loc[
    df["Bilanz Position"].isin(required_positions)
].copy()

# Prepare the data for calculation
quota_rows = []

for product, domestic_use_position in domestic_use_positions.items():
    product_data = quota_input[
        quota_input["Bilanz Produkt"] == product
    ]

    for year in product_data["Wirtschaftsjahr"].unique():
        year_data = product_data[
            product_data["Wirtschaftsjahr"] == year
        ]

        values = year_data.set_index("Bilanz Position")["Bilanz Wert"]

        # Retrieve the required values
        imports = values.get("Einfuhr")
        exports = values.get("Ausfuhr")
        domestic_use = values.get(domestic_use_position)
        usable_production = values.get("Verwendbare Erzeugung")

        # Skip calculations if required values are missing
        if (
            imports is None
            or exports is None
            or domestic_use is None
            or usable_production is None
            or domestic_use == 0
            or usable_production == 0
        ):
            continue

        # Calculate import and export ratios
        import_ratio = imports / domestic_use * 100
        export_ratio = exports / usable_production * 100

        # Store the calculated ratios as new rows
        for position, value in [
            ("Importquote", import_ratio),
            ("Exportquote", export_ratio),
        ]:
            quota_rows.append({
                "Wirtschaftsjahr": year,
                "Bilanz Produkt": product,
                "Bilanz Oberbegriff": "Abgeleitete Kennzahl",
                "Bilanz Position": position,
                "Bilanz Einheit": "%",
                "Bilanz Wert": value,
            })

# Create a separate DataFrame containing the calculated ratios
quota_results = pd.DataFrame(quota_rows)

# Append the new rows without modifying the original dataset
df_extended = pd.concat(
    [df, quota_results],
    ignore_index=True
)

# Display the results
display(quota_results.head(10))

print("Calculated ratio rows:", len(quota_results))
print("Original dataset shape:", df.shape)
print("Extended dataset shape:", df_extended.shape)
```

### Net Trade Balance

The net trade balance is calculated as exports minus imports for each product and financial year.

**Formula:**

Net Trade Balance = Exports − Imports

A positive value indicates that exports exceed imports, while a negative value indicates that imports exceed exports.

The indicator is measured in 1,000 tonnes of product weight. The calculation is performed in a separate DataFrame, leaving the processed dataset unchanged.

#### Calculate Year-over-Year Change in Usable Production

The year-over-year percentage change is calculated separately for barley, potatoes and soft wheat. The percentage change compares each financial year's usable production with the previous financial year's value.

The first financial year has no previous-year comparison and is therefore assigned a missing value. The calculation is performed in a separate DataFrame, leaving the original dataset unchanged.

```python
# Calculate year-over-year changes in usable production

# Sort records by product and financial year
production_change = production_check.sort_values(
    ["Bilanz Produkt", "Wirtschaftsjahr"]
).copy()

# Calculate the previous year's usable production
production_change["Vorjahr"] = (
    production_change.groupby("Bilanz Produkt")["Bilanz Wert"]
    .shift(1)
)

# Calculate the percentage change compared with the previous year
production_change["Veränderung in %"] = (
    (
        production_change["Bilanz Wert"]
        - production_change["Vorjahr"]
    )
    / production_change["Vorjahr"]
    * 100
)

# Select the columns needed for the analysis
production_change = production_change[
    [
        "Wirtschaftsjahr",
        "Bilanz Produkt",
        "Bilanz Wert",
        "Vorjahr",
        "Veränderung in %",
    ]
].copy()

display(production_change.head(10))

print("Total records:", len(production_change))
print(
    "Missing year-over-year changes:",
    production_change["Veränderung in %"].isna().sum()
)
```

```python
# Validate year-over-year changes in usable production

# Summarize the year-over-year changes by product
production_change_validation = production_change.groupby(
    "Bilanz Produkt"
).agg(
    total_rows=("Veränderung in %", "size"),
    missing_changes=(
        "Veränderung in %",
        lambda values: values.isna().sum()
    ),
    minimum_change=("Veränderung in %", "min"),
    maximum_change=("Veränderung in %", "max"),
)

display(production_change_validation.round(2))

# Count missing changes across all products
print(
    "Total missing year-over-year changes:",
    production_change["Veränderung in %"].isna().sum()
)

# Count missing changes for each product
print("\nMissing changes by product:")
display(
    production_change.groupby("Bilanz Produkt")[
        "Veränderung in %"
    ].apply(lambda values: values.isna().sum())
)
```

```python
# Summarize year-over-year changes by product

production_change_summary = production_change.groupby(
    "Bilanz Produkt"
)["Veränderung in %"].agg(
    minimum_change="min",
    maximum_change="max",
    mean_change="mean",
    mean_absolute_change=lambda values: values.abs().mean(),
)

display(production_change_summary.round(2))
```

The year-over-year percentage change in domestic use measures how the domestic use of barley, potatoes and soft wheat changes compared with the previous financial year.

For barley and soft wheat, the balance item "Inlandsverwendung insgesamt" is used. For potatoes, the balance item "Inlandsverwendung" is used, following the previously established methodology.

Changes in domestic use can help contextualize developments in import ratios and self-sufficiency rates. However, this indicator alone does not explain the causes of changes in domestic use.
