```python

import pandas as pd
from pathlib import Path

# Define the directory
data_dir = Path(
    r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\processed"
)

# Define the file paths
files = {
    "EU": data_dir / "bmelh_production_prices_eu_2009-2024.xlsx",
    "Germany": data_dir / "bmelh_production_prices_germany_2009-2024.xlsx"
}

# Load both Excel files
prices_eu = pd.read_excel(files["EU"])
prices_germany = pd.read_excel(files["Germany"])

# Check whether both files exist
for country, file_path in files.items():
    print(f"{country} file exists:", file_path.exists())

# Inspect both datasets
datasets = {
    "EU": prices_eu,
    "Germany": prices_germany
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(f"{name} production prices")
    print(f"{'=' * 50}")

    # Check dataset dimensions
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    # Check column names
    print("\nColumn names:")
    print(df.columns.tolist())

    # Check duplicate rows
    print("\nDuplicate rows:", df.duplicated().sum())

    # Check missing values
    print("\nMissing values per column:")
    print(df.isna().sum())

    # Display sample rows
    print("\nFirst five rows:")
    display(df.head())

    # Display data types
    print("\nData types:")
    print(df.dtypes)
```

```python

# Check placeholder values without modifying the datasets
datasets = {
    "EU": prices_eu,
    "Germany": prices_germany
}

placeholders = [".", "-", "..", "..."]

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(f"{name}: Placeholder inspection")
    print(f"{'=' * 50}")

    # Count placeholders in each column
    for column in df.columns:
        values = df[column].astype("string").str.strip()
        mask = values.isin(placeholders)

        if mask.any():
            print(f"\nColumn: {column}")
            print("Placeholder counts:")
            print(values[mask].value_counts().to_dict())

            # Display affected rows
            display(df.loc[mask])
```

```python

# Summarize dot placeholders by year
datasets = {
    "EU": prices_eu,
    "Germany": prices_germany
}

for name, df in datasets.items():
    year_columns = [
        col for col in df.columns
        if str(col).isdigit() and 2009 <= int(col) <= 2024
    ]

    dot_counts = df[year_columns].apply(
        lambda col: col.astype("string").str.strip().eq(".").sum()
    )

    print(f"\n{name}: missing-price placeholders by year")
    print(dot_counts.to_string())
    print("Total placeholders:", int(dot_counts.sum()))
```

```python

# Create analysis copies and convert placeholders to missing values
import pandas as pd

placeholders = [".", "-", "--"]

prices_eu_clean = prices_eu.copy()
prices_germany_clean = prices_germany.copy()

for df in [prices_eu_clean, prices_germany_clean]:
    year_columns = [
        col for col in df.columns
        if str(col).isdigit() and 2009 <= int(col) <= 2024
    ]

    for col in year_columns:
        values = df[col].astype("string").str.strip()
        df[col] = values.mask(values.isin(placeholders), pd.NA)
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Report missing values in year columns
for name, df in {
    "EU": prices_eu_clean,
    "Germany": prices_germany_clean
}.items():
    year_columns = [
        col for col in df.columns
        if str(col).isdigit() and 2009 <= int(col) <= 2024
    ]

    print(f"\n{name}: missing values by year")
    print(df[year_columns].isna().sum().to_string())
    print("Total missing values:", int(df[year_columns].isna().sum().sum()))
```

```python

# Identify rows containing missing prices

def show_missing_rows(df, dataset_name):
    year_columns = [
        col for col in df.columns
        if str(col).isdigit() and 2009 <= int(col) <= 2024
    ]

    missing_mask = df[year_columns].isna()
    result = df.loc[
        missing_mask.any(axis=1),
        [col for col in df.columns if col not in year_columns]
    ].copy()

    result["Missing years"] = missing_mask.loc[
        missing_mask.any(axis=1)
    ].apply(
        lambda row: ", ".join(str(year) for year in row.index[row].tolist()),
        axis=1
    )

    result["Number missing"] = missing_mask.sum(axis=1).loc[
        missing_mask.any(axis=1)
    ].values

    print(f"\n{dataset_name}: rows with missing prices")
    print(result.to_string(index=False))


show_missing_rows(prices_eu_clean, "EU")
show_missing_rows(prices_germany_clean, "Germany")
```

```python

# Summarize missing prices by country and product

year_columns = [
    col for col in prices_eu_clean.columns
    if str(col).isdigit() and 2009 <= int(col) <= 2024
]

eu_with_missing_count = prices_eu_clean.assign(
    Missing=prices_eu_clean[year_columns].isna().sum(axis=1)
)

country_summary = (
    eu_with_missing_count
    .groupby("Land", dropna=False)["Missing"]
    .agg(
        Total_missing="sum",
        Product_series="size",
        Max_missing_in_one_series="max"
    )
    .sort_values("Total_missing", ascending=False)
)

product_summary = (
    eu_with_missing_count
    .groupby("landwirtschaftliches Produkt", dropna=False)["Missing"]
    .agg(
        Total_missing="sum",
        Country_series="size",
        Max_missing_in_one_series="max"
    )
    .sort_values("Total_missing", ascending=False)
)

print("EU: Missing values by country")
print(country_summary.to_string())

print("\nEU: Missing values by product")
print(product_summary.to_string())
```

```python

# Calculate data completeness for each country-product series

year_columns = [
    col for col in prices_eu_clean.columns
    if str(col).isdigit() and 2009 <= int(col) <= 2024
]

series_quality = prices_eu_clean[
    ["Land", "landwirtschaftliches Produkt"] + year_columns
].copy()

series_quality["Missing years"] = (
    series_quality[year_columns].isna().sum(axis=1)
)

series_quality["Available years"] = (
    len(year_columns) - series_quality["Missing years"]
)

series_quality["Completeness (%)"] = (
    series_quality["Available years"] / len(year_columns) * 100
).round(1)

series_quality = series_quality[
    [
        "Land",
        "landwirtschaftliches Produkt",
        "Available years",
        "Missing years",
        "Completeness (%)"
    ]
].sort_values(
    ["Completeness (%)", "Land"],
    ascending=[True, True]
)

# Display all country-product series, starting with the least complete
display(series_quality)
```

```python

# Select candidate series with at least 12 available years

year_columns = [
    col for col in prices_eu_clean.columns
    if str(col).isdigit() and 2009 <= int(col) <= 2024
]

series_detail = prices_eu_clean[
    ["Land", "landwirtschaftliches Produkt"] + year_columns
].copy()

series_detail["Available years"] = series_detail[year_columns].notna().sum(axis=1)
series_detail["Missing years"] = series_detail[year_columns].apply(
    lambda row: ", ".join(
        str(year) for year in year_columns if pd.isna(row[year])
    ),
    axis=1
)

# Keep series with at least 75% data coverage
candidates = series_detail[
    series_detail["Available years"] >= 12
].copy()

candidates = candidates[
    [
        "Land",
        "landwirtschaftliches Produkt",
        "Available years",
        "Missing years"
    ]
].sort_values(
    ["landwirtschaftliches Produkt", "Land"]
)

print(f"Suitable candidates: {len(candidates)} of {len(series_detail)} series")
display(candidates)

print("\nCandidate counts by product:")
display(
    candidates.groupby("landwirtschaftliches Produkt")
    .size()
    .rename("Number of candidate series")
    .to_frame()
)
```

```python

# Identify common comparison years for each product

year_columns = [
    col for col in prices_eu_clean.columns
    if str(col).isdigit() and 2009 <= int(col) <= 2024
]

candidate_rows = series_detail[
    series_detail["Available years"] >= 12
].copy()

product_column = "landwirtschaftliches Produkt"

for product in sorted(candidate_rows[product_column].dropna().unique()):
    subset = candidate_rows[
        candidate_rows[product_column] == product
    ]

    # Count available country series for each year
    available_counts = subset[year_columns].notna().sum(axis=0)

    coverage_table = pd.DataFrame({
        "Available country series": available_counts,
        "Coverage (%)": (
            available_counts / len(subset) * 100
        ).round(1)
    })

    # Find years available for every candidate series
    common_years = available_counts[
        available_counts == len(subset)
    ].index.tolist()

    print(f"\n{'=' * 50}")
    print(f"Product: {product}")
    print(f"Candidate series: {len(subset)}")
    print(f"Common years: {common_years}")
    print("\nYearly coverage:")
    display(coverage_table)
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Select the year columns from the German dataset
year_columns = [
    col for col in prices_germany_clean.columns
    if str(col).isdigit() and 2009 <= int(col) <= 2024
]

# Reshape the German data into long format
germany_long = prices_germany_clean.melt(
    id_vars=["landwirtschaftliches Produkt"],
    value_vars=year_columns,
    var_name="Jahr",
    value_name="Erzeugerpreis"
)

germany_long["Jahr"] = pd.to_numeric(germany_long["Jahr"])
germany_long["Erzeugerpreis"] = pd.to_numeric(
    germany_long["Erzeugerpreis"],
    errors="coerce"
)

germany_long = germany_long.sort_values("Jahr")

# Summarize the price development for each product
summary = (
    germany_long.dropna(subset=["Erzeugerpreis"])
    .groupby("landwirtschaftliches Produkt")
    .agg(
        Erstes_Jahr=("Jahr", "min"),
        Letztes_Jahr=("Jahr", "max"),
        Verfügbare_Jahre=("Erzeugerpreis", "count"),
        Durchschnitt=("Erzeugerpreis", "mean"),
        Minimum=("Erzeugerpreis", "min"),
        Maximum=("Erzeugerpreis", "max"),
        Anfangspreis=("Erzeugerpreis", "first"),
        Letzter_Preis=("Erzeugerpreis", "last")
    )
)

summary["Preisänderung"] = (
    summary["Letzter_Preis"] - summary["Anfangspreis"]
)

summary["Preisänderung (%)"] = (
    (summary["Letzter_Preis"] / summary["Anfangspreis"] - 1) * 100
).round(1)

summary = summary.round(2)

print("German agricultural producer prices, 2009–2024")
display(summary)

# Plot one price series per product
products = germany_long["landwirtschaftliches Produkt"].dropna().unique()

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)
axes = axes.flatten()

for ax, product in zip(axes, products):
    subset = germany_long[
        germany_long["landwirtschaftliches Produkt"] == product
    ].dropna(subset=["Erzeugerpreis"])

    ax.plot(subset["Jahr"], subset["Erzeugerpreis"], marker="o")
    ax.set_title(product)
    ax.set_ylabel("€ per 100 kg")
    ax.set_xticks([2009, 2012, 2015, 2018, 2021, 2024])
    ax.grid(True, alpha=0.3)

plt.suptitle("German producer prices by agricultural product")
plt.tight_layout()
plt.show()
```
