```python

import pandas as pd
from pathlib import Path

data_path = Path(
    r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\processed"
)

# Dateien laden
df_final = pd.read_excel(
    data_path / "ble_supply_balance_final_1991-2025.xlsx"
)

df_net = pd.read_excel(
    data_path / "ble_net_trade_balance_1991-2025.xlsx"
)

keys = ["Wirtschaftsjahr", "Bilanz Produkt"]

# Finale Tabelle mit der Originaldatei vergleichen
check = df_final.merge(
    df_net[keys + ["Nettohandelsbilanz", "Einfuhr", "Ausfuhr"]],
    on=keys,
    how="outer",
    suffixes=("_final", "_original"),
    indicator=True,
    validate="one_to_one"
)

print("=== Merge-Prüfung ===")
print(check["_merge"].value_counts())

print("\n=== Abweichungen ===")

for col in ["Nettohandelsbilanz", "Einfuhr", "Ausfuhr"]:
    final_col = f"{col}_final"
    original_col = f"{col}_original"

    # Numerischer Vergleich, fehlende Werte werden berücksichtigt
    a = pd.to_numeric(check[final_col], errors="coerce")
    b = pd.to_numeric(check[original_col], errors="coerce")

    equal = a.eq(b) | (a.isna() & b.isna())

    print(f"{col}: {int((~equal).sum())} Abweichungen")

    if (~equal).any():
        print(check.loc[
            ~equal,
            keys + [final_col, original_col]
        ].head(10))

print("\n=== Gesamtprüfung ===")
print("Zeilen finale Datei:", len(df_final))
print("Zeilen Original Net Trade:", len(df_net))
print("Zeilen im Vergleich:", len(check))
```

```python

import pandas as pd
from pathlib import Path
import numpy as np

data_path = Path(
    r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\processed"
)

# Dateien laden
df_final = pd.read_excel(
    data_path / "ble_supply_balance_final_1991-2025.xlsx"
)

df_domestic = pd.read_excel(
    data_path / "ble_domestic_use_change_1991-2025.xlsx"
)

# Schlüssel und zu vergleichende Spalten
keys = ["Wirtschaftsjahr", "Bilanz Produkt"]
col_final = "Inlandsverwendung standardisiert"
col_domestic = "Inlandsverwendung geprüft"

# Prüfen, ob alle benötigten Spalten existieren
for df, cols, name in [
    (df_final, keys + [col_final], "Final"),
    (df_domestic, keys + [col_domestic], "Domestic Use Change")
]:
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise ValueError(f"{name}: Fehlende Spalten: {missing}")

# Doppelte Schlüssel prüfen
for df, name in [
    (df_final, "Final"),
    (df_domestic, "Domestic Use Change")
]:
    duplicates = df.duplicated(subset=keys, keep=False)
    print(f"{name} – Zeilen mit doppelten Schlüsseln: {duplicates.sum()}")

# Nur benötigte Spalten auswählen und zusammenführen
check = df_final[keys + [col_final]].merge(
    df_domestic[keys + [col_domestic]],
    on=keys,
    how="outer",
    indicator=True,
    validate="one_to_one"
)

# Vergleich numerisch, mit Toleranz für Rundungsdifferenzen
a = pd.to_numeric(check[col_final], errors="coerce")
b = pd.to_numeric(check[col_domestic], errors="coerce")

both_missing = a.isna() & b.isna()
both_present = a.notna() & b.notna()

equal = both_missing | (
    both_present & np.isclose(a, b, rtol=1e-9, atol=1e-8)
)

matched_keys = check["_merge"].eq("both")
equal_for_matched = equal & matched_keys

print("\n=== Schlüsselabgleich ===")
print(check["_merge"].value_counts())

print("\n=== Wertevergleich ===")
print("Gemeinsame Schlüssel:", matched_keys.sum())
print("Übereinstimmende Werte:", equal_for_matched.sum())
print(
    "Abweichende Werte:",
    (matched_keys & ~equal).sum()
)
print(
    "Fehlende Werte in mindestens einer Datei:",
    (matched_keys & (a.isna() ^ b.isna())).sum()
)

# Abweichungen ausgeben
mismatches = check.loc[
    matched_keys & ~equal,
    keys + [col_final, col_domestic]
]

if not mismatches.empty:
    print("\nErste 20 Abweichungen:")
    print(mismatches.head(20).to_string(index=False))
else:
    print("\nAlle Werte bei gemeinsamen Schlüsseln stimmen überein.")

# Schlüssel, die nur in einer Datei vorkommen
unmatched = check.loc[~matched_keys, keys + ["_merge"]]
if not unmatched.empty:
    print("\nSchlüssel, die nicht in beiden Dateien vorkommen:")
    print(unmatched.to_string(index=False))
```

```python

import pandas as pd
from pathlib import Path

# ============================================================
# 1. Dateipfade
# ============================================================

data_dir = Path(
    r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\processed"
)

final_file = data_dir / "ble_supply_balance_final_1991-2025.xlsx"
source_file = data_dir / "ble_domestic_use_change_1991-2025.xlsx"

# ============================================================
# 2. Dateien laden
# ============================================================

df_final = pd.read_excel(final_file)
df_source = pd.read_excel(source_file)

keys = ["Wirtschaftsjahr", "Bilanz Produkt"]

source_cols = [
    "Inlandsverwendung geprüft",
    "Vorjahr",
    "Veränderung in %"
]

rename_cols = {
    "Inlandsverwendung geprüft": "Inlandsverwendung_geprueft",
    "Vorjahr": "Inlandsverwendung_Vorjahr",
    "Veränderung in %": "Inlandsverwendung_Veraenderung_%"
}

# ============================================================
# 3. Spalten prüfen
# ============================================================

for col in keys:
    if col not in df_final.columns:
        raise ValueError(f"Schlüsselspalte fehlt in Finaldatei: {col}")
    if col not in df_source.columns:
        raise ValueError(f"Schlüsselspalte fehlt in Quelldatei: {col}")

missing_source_cols = [
    col for col in source_cols
    if col not in df_source.columns
]

if missing_source_cols:
    raise ValueError(
        f"Fehlende Spalten in Quelldatei: {missing_source_cols}"
    )

# ============================================================
# 4. Schlüssel auf Eindeutigkeit prüfen
# ============================================================

if df_final.duplicated(keys).any():
    raise ValueError("Die Finaldatei enthält doppelte Schlüssel.")

if df_source.duplicated(keys).any():
    raise ValueError("Die Quelldatei enthält doppelte Schlüssel.")

# ============================================================
# 5. Drei Quellspalten vorbereiten
# ============================================================

source_data = df_source[
    keys + source_cols
].rename(columns=rename_cols)

# Bereits vorhandene Zielspalten aus der Finaldatei entfernen,
# damit beim Merge keine _x- und _y-Spalten entstehen.
target_cols = list(rename_cols.values())

df_final_base = df_final.drop(
    columns=[
        col for col in target_cols
        if col in df_final.columns
    ]
)

# ============================================================
# 6. Merge durchführen
# ============================================================

df_merged = df_final_base.merge(
    source_data,
    on=keys,
    how="left",
    validate="one_to_one",
    indicator=True
)

print("=== Merge-Ergebnis ===")
print("Zeilen vorher:", len(df_final))
print("Zeilen nachher:", len(df_merged))
print("Zuordnungen:", int(df_merged["_merge"].eq("both").sum()))
print("Ohne Quelldatensatz:", int(df_merged["_merge"].eq("left_only").sum()))

# Nicht zugeordnete Datensätze anzeigen
unmatched = df_merged["_merge"].eq("left_only")

if unmatched.any():
    print("\nNicht zugeordnete Schlüssel:")
    print(
        df_merged.loc[unmatched, keys]
        .to_string(index=False)
    )

# Merge-Indikator entfernen
df_merged = df_merged.drop(columns="_merge")

# ============================================================
# 7. Ergebnis kontrollieren
# ============================================================

print("\n=== Neue Spalten ===")
for col in target_cols:
    print(f"{col}: vorhanden = {col in df_merged.columns}")

print("\nDimensionen vorher:", df_final.shape)
print("Dimensionen nachher:", df_merged.shape)

suffix_cols = [
    col for col in df_merged.columns
    if col.endswith(("_x", "_y"))
]

print("Spalten mit _x oder _y:", suffix_cols)

if len(df_merged) != len(df_final):
    raise ValueError(
        "Die Zeilenzahl hat sich verändert. Datei wird nicht gespeichert."
    )

if not all(col in df_merged.columns for col in target_cols):
    raise ValueError(
        "Mindestens eine der drei Zielspalten fehlt. Datei wird nicht gespeichert."
    )

if unmatched.any():
    raise ValueError(
        "Nicht alle Datensätze konnten zugeordnet werden. "
        "Bitte Schlüssel prüfen. Datei wurde nicht gespeichert."
    )

# ============================================================
# 8. Finaldatei speichern
# ============================================================

df_merged.to_excel(final_file, index=False)

print("\nFinaldatei erfolgreich gespeichert:")
print(final_file)
print("\n=== Merge abgeschlossen ===")
```

```python

import pandas as pd
from pathlib import Path

# --------------------------------------------------
# 1. Dateien laden
# --------------------------------------------------
final_file = Path(final_file)

production_file = (
    final_file.parent / "ble_production_change_1991-2025.xlsx"
)

df_final = pd.read_excel(final_file)
df_prod = pd.read_excel(production_file)

# Spaltennamen vereinheitlichen: Leerzeichen -> Unterstriche
for df in [df_final, df_prod]:
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.replace(r"\s+", "_", regex=True)
    )

print("Finaldatei:", df_final.shape)
print("Production-Change-Datei:", df_prod.shape)

# --------------------------------------------------
# 2. Schlüssel prüfen
# --------------------------------------------------
keys = ["Wirtschaftsjahr", "Bilanz_Produkt"]

for key in keys:
    if key not in df_final.columns:
        raise KeyError(f"Schlüssel fehlt in Finaldatei: {key}")
    if key not in df_prod.columns:
        raise KeyError(f"Schlüssel fehlt in Production-Datei: {key}")

# Die letzten beiden Spalten der Production-Datei
source_cols = list(df_prod.columns[-2:])

if any(col in keys for col in source_cols):
    raise ValueError(
        f"Die letzten beiden Spalten sind nicht beide Daten-Spalten: "
        f"{source_cols}"
    )

print("\nLetzte zwei Spalten der Production-Datei:")
print(source_cols)

# --------------------------------------------------
# 3. Spalten auswählen und umbenennen
# --------------------------------------------------
rename_map = {
    source_cols[0]: "Verwendbare_Erzeugung_Vorjahr",
    source_cols[1]: "Verwendbare_Erzeugung_Veraenderung_%"
}

df_prod_subset = df_prod[keys + source_cols].rename(
    columns=rename_map
)

target_cols = list(rename_map.values())

# Doppelte Schlüssel verhindern
if df_final.duplicated(keys).any():
    raise ValueError("Die Finaldatei enthält doppelte Schlüssel.")

if df_prod_subset.duplicated(keys).any():
    raise ValueError(
        "Die Production-Datei enthält doppelte Schlüssel."
    )

# Vorhandene Zielspalten ersetzen, falls sie bereits existieren
df_final = df_final.drop(
    columns=target_cols,
    errors="ignore"
)

# --------------------------------------------------
# 4. Merge durchführen
# --------------------------------------------------
df_merged = df_final.merge(
    df_prod_subset,
    on=keys,
    how="left",
    validate="one_to_one",
    indicator=True
)

matched = (df_merged["_merge"] == "both").sum()
unmatched = (df_merged["_merge"] == "left_only").sum()

print("\n=== Merge-Ergebnis ===")
print("Zeilen vorher:", len(df_final))
print("Zeilen nachher:", len(df_merged))
print("Zuordnungen:", matched)
print("Ohne Quelldatensatz:", unmatched)

# Nicht speichern, wenn Zuordnungen fehlen oder Zeilenanzahl abweicht
if len(df_merged) != len(df_final):
    raise ValueError("Die Zeilenanzahl hat sich verändert.")

if unmatched > 0:
    print("\nNicht zugeordnete Datensätze:")
    display(
        df_merged.loc[
            df_merged["_merge"] == "left_only",
            keys
        ]
    )
    raise ValueError(
        "Es fehlen Zuordnungen. Die Finaldatei wurde nicht gespeichert."
    )

# --------------------------------------------------
# 5. Ergebnis prüfen und speichern
# --------------------------------------------------
df_merged = df_merged.drop(columns="_merge")

print("\nNeue Spalten:")
display(df_merged[keys + target_cols].head(10))

df_merged.to_excel(final_file, index=False)

print("\nFinaldatei erfolgreich aktualisiert:")
print(final_file)
print("Neue Dimensionen:", df_merged.shape)
```
