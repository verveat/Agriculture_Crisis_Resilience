```python
import pandas as pd

# Dateipfad
file_path = r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\processed\ble_cleaned_supply_balance_with_quotas_heute_1991-2025.xlsx"

# Excel-Datei einlesen
df_with_quotas = pd.read_excel(file_path)

# _x und _y am Ende der Spaltennamen entfernen
df_with_quotas.columns = df_with_quotas.columns.str.replace(
    r'_[xy]$', '', regex=True
)

# Erste Ergebnisse anzeigen
display(df_with_quotas.head())

# Datensatz überprüfen
print("Anzahl Zeilen:", len(df_with_quotas))
print("Anzahl Spalten:", len(df_with_quotas.columns))

print("\nBereinigte Spaltennamen:")
print(df_with_quotas.columns.tolist())

print("\nDatentypen:")
print(df_with_quotas.dtypes)

# Prüfen, ob doppelte Spaltennamen entstanden sind
print("\nDoppelte Spaltennamen:")
print(df_with_quotas.columns[df_with_quotas.columns.duplicated()].tolist())
```

```python

# Bereinigte Excel-Datei speichern
output_path = r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\processed\ble_cleaned_supply_balance_food_use_1991-2025.xlsx"

df_with_quotas.to_excel(output_path, index=False)

print("Datei erfolgreich gespeichert unter:")
print(output_path)
```
