```python

from pathlib import Path
import pandas as pd

# ==================================================
# 1. Pfad zum Datenverzeichnis
# ==================================================

data_path = Path(
    r"C:\Users\incog\Desktop\Portfolio_Project_Agriculture\data\raw\destatis_aussenhandel_2008-2026_weizen_und_mengkorn_gerste_kartoffel"
)

# ==================================================
# 2. Dateinamen definieren
# ==================================================

file_weizen_mengkorn = (
    "destatis_51000-0009_aussenhandel_ein_ausfuhr_"
    "weizen_mengkorn_2008-2025.xls"
)

file_gerste = (
    "destatis_51000-0009_aussenhandel_ein-ausfuhr_"
    "gerste_2008-2025.xls"
)

file_kartoffel = (
    "destatis_51000-0009_aussenhandel_ein-ausfuhr_"
    "kartoffel_2008_2025.xls"
)

# ==================================================
# 3. Excel-Dateien einlesen
# ==================================================

path_weizen_mengkorn = data_path / file_weizen_mengkorn
path_gerste = data_path / file_gerste
path_kartoffel = data_path / file_kartoffel

# Prüfen, ob alle Dateien vorhanden sind
for file_path in [
    path_weizen_mengkorn,
    path_gerste,
    path_kartoffel
]:
    if not file_path.exists():
        raise FileNotFoundError(
            f"Datei nicht gefunden: {file_path}"
        )

# Excel-Dateien laden
df_weizen_mengkorn = pd.read_excel(
    path_weizen_mengkorn
)

df_gerste = pd.read_excel(
    path_gerste
)

df_kartoffel = pd.read_excel(
    path_kartoffel
)

# ==================================================
# 4. Daten überprüfen
# ==================================================

dataframes = {
    "Weizen und Mengkorn": df_weizen_mengkorn,
    "Gerste": df_gerste,
    "Kartoffeln": df_kartoffel
}

for name, df in dataframes.items():
    print(f"\n{'=' * 60}")
    print(name)
    print("=" * 60)

    print(f"Anzahl Zeilen: {df.shape[0]}")
    print(f"Anzahl Spalten: {df.shape[1]}")

    print("\nSpaltennamen:")
    print(df.columns.tolist())

    print("\nErste fünf Zeilen:")
    print(df.head())

# ==================================================
# 5. Optional: Datentypen überprüfen
# ==================================================

print("\nDatentypen – Weizen und Mengkorn:")
print(df_weizen_mengkorn.dtypes)

print("\nDatentypen – Gerste:")
print(df_gerste.dtypes)

print("\nDatentypen – Kartoffeln:")
print(df_kartoffel.dtypes)
```
