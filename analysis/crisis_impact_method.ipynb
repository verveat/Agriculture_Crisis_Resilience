## Step 8: Historical Crisis Chronology and Temporal Classification

### 8.1 Objective

This step establishes a historically grounded chronology of major crisis events within and around the observation period of this study (1991/92–2024/25).

The objective is to identify and document the historical beginning and, where applicable, the end of each crisis at monthly resolution.

At this stage, the analysis focuses exclusively on the historical timing and duration of crisis events. No assumptions are made about their potential effects on German agricultural markets, individual agricultural products, agricultural production, domestic use, or self-sufficiency.

The crisis chronology is established independently of the empirical results. This prevents crisis periods from being defined retrospectively based on observed changes in the agricultural balance data.

### 8.2 Methodological Approach to Crisis Dating

Two complementary approaches are applied to determine the monthly start and end dates of crisis events.

**1. Source-based dating**

Source-based dating is applied when reliable historical, scientific, or institutional sources document a specific event or milestone with an identifiable date.

The corresponding calendar month is adopted as the historical reference month. Examples include legally effective political events, formally documented decisions, and dated military or institutional developments.

The source must support the specific historical event or milestone used to establish the date. A general description of a crisis is not sufficient to verify a specific starting month.

**2. Operational dating**

Operational dating is applied when a broader crisis does not have an unambiguous historical starting date at monthly resolution.

The operational date is derived using a clearly documented historical reference criterion. Depending on the nature of the event, this may involve:

- monthly historical indicators, such as precipitation, drought, energy prices, fertiliser prices, or inflation rates;
- a dated historical event used as an operational proxy for the beginning of a broader crisis;
- the beginning of a defined observation interval used to examine a subsequent development.

The selected data series or historical reference event, the decision criterion, and the derivation of the assigned month must be documented.

An operational reference month must not automatically be interpreted as the objectively established beginning of the entire crisis. In particular, the beginning of an observation interval may differ from the onset of the underlying historical developments.

Operational dates are therefore distinguished from directly documented historical dates and are interpreted according to their specific methodological purpose.

**Source documentation**

Priority is given to primary sources and publications from recognized institutions, including the German Weather Service (DWD), the Food and Agriculture Organization (FAO), the World Health Organization (WHO), the European Commission, Eurostat, the International Maritime Organization (IMO), and the United Nations.

Where several sources are required to support different aspects of an event, all relevant references are documented.

### 8.3 Documentation and Classification of Crisis Events

Each crisis is recorded using a standardized set of variables:

- `Event_ID`: Unique identifier of the crisis event.
- `Event_Name`: Historical designation of the crisis.
- `Period_Original`: Original broad historical period associated with the event. This field does not necessarily represent a verified start and end date.
- `Start_Month`: Historical or operationally determined starting month in YYYY-MM format.
- `End_Month`: Historical or operationally determined ending month in YYYY-MM format, if established.
- `Dating_Method`: Indicates whether source-based dating or operational dating is applied.
- `Dating_Status`: Indicates the degree to which the assigned starting month is supported by the available evidence.
- `End_Status`: Indicates whether the end of the event has been established, remains undetermined, is pending determination, or is documented as ongoing.
- `Dating_Evidence`: Summary of the historical evidence supporting the assigned date.
- `Source`: Historical, scientific, or institutional references supporting the event description and dating.
- `Operationalization_Rule`: Description of the criterion used to establish the date, or an explanation of why no operationalization is required.

The following categories are used consistently:

**Dating_Status**

- `Verified`: The specific historical date or month is directly documented by a reliable source.
- `Source-supported`: The assigned month is supported by historical evidence but involves some interpretation or an operational reference criterion.
- `Preliminary`: The assigned month is retained as a provisional reference because the source documentation or operational criterion is not yet sufficiently established.

**End_Status**

- `Undetermined`: No defensible end month has yet been established.
- `Pending determination`: A specific procedure for determining the end has been defined, but the end date has not yet been established.
- `Ongoing`: The event is documented as continuing at the time of the latest available assessment.

If the end of a crisis cannot be established, `End_Month` remains empty. The end of the BLE observation period must not be interpreted as the historical end of a crisis.

### 8.4 Distinguishing Historical Crisis Dates from Empirical Observations

The BLE supply balance data are reported by agricultural year (*Wirtschaftsjahr*), which runs from 1 July to 30 June of the following calendar year.

Consequently, historically documented crisis months can be assigned to the corresponding agricultural year or years. This allows the analysis to identify which annual observations overlap with a given crisis period.

The historical crisis chronology and the empirical analysis serve different purposes:

- Historical sources and operational definitions establish the preliminary monthly beginning and, where possible, the end of crisis events.
- These monthly dates are subsequently mapped to the corresponding BLE agricultural years.
- The BLE data are used to examine annual changes in the selected agricultural indicators across the agricultural years overlapping with the respective crisis periods.

Since the BLE data provide annual rather than monthly observations, changes in the selected indicators can only be identified at the agricultural-year level. The exact month in which an observed change begins or ends within an agricultural year cannot be determined from these data alone.

Events that fall partly or entirely outside the BLE observation period may remain part of the historical chronology. However, they cannot be treated as fully observable empirical crisis episodes within the BLE dataset.

The comparison therefore examines the temporal consistency between historically identified crisis periods and observed changes in the agricultural indicators. It does not assume that historical crisis dates and empirical changes necessarily coincide.

### 8.5 Separation of Historical Dating and Empirical Interpretation

The historical crisis dates established in this step serve as reference points for the subsequent empirical analysis.

They are not treated as definitive evidence of when a crisis began to affect German agricultural markets or when its effects ended.

The subsequent analysis will investigate:

- whether changes in the selected agricultural indicators occur during agricultural years overlapping with historically identified crisis periods;
- whether observable changes emerge in subsequent agricultural years;
- how long deviations from the respective pre-crisis reference levels persist;
- whether the observed patterns are consistent with the historical chronology.

The empirical results may provide indications that the timing or duration of observable agricultural-market changes differs from the historically documented crisis period.

However, the annual BLE data alone cannot establish an exact monthly onset or end of an agricultural-market response. Furthermore, temporal coincidence does not by itself demonstrate a causal relationship between a crisis and an observed change.

### 8.6 Methodological Rationale

This approach ensures a clear separation between historical event classification and empirical outcome analysis.

The specific dating procedure is adapted to the historical characteristics of each event. Identical indicators or thresholds are not imposed on fundamentally different types of crises.

Methodological consistency is instead ensured through common documentation requirements: every assigned month must have an identifiable historical or operational basis, a transparent decision criterion, an appropriate source, and an explicit assessment of remaining uncertainty.

Where a historical month is directly documented, no additional operationalization is required. Where operational dating is applied, the chosen indicator, event proxy, or observation interval must be distinguished from the broader historical crisis itself.

This approach provides a reproducible basis for comparing different crisis events while avoiding unsupported or arbitrary date assignments. It also prevents crisis periods from being defined based on the empirical results that they are subsequently intended to explain.

## 9. Empirical Analysis of Agricultural Market Resilience

### 9.1 Objective

The objective of this section is to examine the development of selected German agricultural market indicators following the historically documented beginning of the identified crisis events.

The analysis builds on the crisis chronology established in Step 8. The twelve documented historical starting months are retained without modification and assigned to the corresponding BLE agricultural years (Wirtschaftsjahre).

For each product and indicator, annual observations are compared with a pre-crisis reference level and a linear pre-crisis trend. This makes it possible to examine deviations from the previous average level as well as deviations from the development observed before the crisis.

The analysis is conducted separately for barley, potatoes and soft wheat.

### 9.2 Assignment of Historical Events to BLE Agricultural Years

The BLE agricultural year runs from 1 July to 30 June. Historical crisis starting months are therefore assigned to the agricultural year in which they occur.

For example, a crisis beginning in February 2022 is assigned to the BLE agricultural year 2021/22, whereas an event beginning in September 2022 is assigned to 2022/23.

The original historical starting month remains stored in the crisis chronology. The assigned agricultural year is used to establish temporal consistency between historical events and the annual BLE observations.

Since the BLE data are annual, the precise month in which a market response occurs cannot be identified. An event beginning during an agricultural year may therefore affect only part of the corresponding annual observation period.

The assigned event year is consequently interpreted as the first potentially affected agricultural year, not as a full year of crisis exposure.

### 9.3 Pre-Crisis Reference Periods and Sensitivity Analysis

The pre-crisis reference level is calculated as the arithmetic mean of the indicator values in the agricultural years preceding the assigned event year.

Two reference periods are examined:

- **Three pre-crisis years:** main specification.
- **Five pre-crisis years:** sensitivity analysis.

The comparison assesses whether the results depend on the length of the reference period. A longer reference period may reduce the influence of individual annual fluctuations, but it may also include structural changes in agricultural markets or statistical data collection.

The five-year reference period is therefore not automatically considered superior to the three-year period.

Only complete pre-crisis reference periods are used. If the required observations are unavailable, the respective calculation is documented as unavailable rather than based on an incomplete or artificially extended reference period.

### 9.4 Annual Deviations from the Reference Level and Pre-Crisis Trend

The analysis calculates annual deviations for the assigned event year and the following ten agricultural years (\(t=0\) to \(t+10\)), depending on data availability.

Two reference measures are used:

1. **Pre-crisis reference level:** the arithmetic mean of the selected pre-crisis observations.
2. **Pre-crisis trend:** a linear trend estimated exclusively from the pre-crisis observations and extrapolated into the subsequent years.

For both measures, absolute and relative deviations are calculated.

The annual event-time observations make it possible to examine the magnitude, direction and persistence of changes rather than relying exclusively on averages across broad periods.

The event year (\(t=0\)) is treated as the first potentially affected agricultural year. The following years are initially classified as follow-up observations. They are not automatically interpreted as a historically established post-crisis period.

### 9.5 Differentiation of the Follow-up Period

To examine the temporal development of agricultural market resilience in greater detail, the follow-up period is divided into three analytical stages:

- **Short-term follow-up period (\(t+1\) to \(t+2\)):** captures the immediate development after the event year and indicates whether the initial deviation persists or begins to diminish.
- **Medium-term follow-up period (\(t+3\) to \(t+5\)):** examines whether the indicator continues to recover, stabilizes or remains deviated from its pre-crisis reference level.
- **Long-term follow-up period (\(t+6\) to \(t+10\)):** examines whether the observed development persists over a longer period, whether the indicator approaches its pre-crisis reference level, or whether a more lasting structural change may have occurred.

The annual observations remain the basis of the analysis. Results are therefore evaluated both individually for each agricultural year and descriptively for the three follow-up stages.

This differentiation represents an analytical time-window definition. It does not imply that every crisis ends after the event year or that the subsequent years automatically constitute a historically established post-crisis period.

The availability of long-term observations depends on the position of each event within the BLE observation period. Incomplete follow-up windows are documented explicitly and are not extrapolated or artificially completed.

Long-term developments must also be interpreted in the context of subsequent crisis events, structural changes and other market influences. They cannot automatically be attributed to the original crisis event.

The availability of follow-up observations differs across crisis events due to the limited observation period of the BLE data. While the short-term and medium-term periods can be examined for several events, the long-term follow-up period is only fully observable for selected earlier events. The long-term analysis is therefore treated as a supplementary analytical perspective. Incomplete observation windows are explicitly identified, and comparisons are restricted accordingly.

### 9.6 Summary Measures of Market Resilience

The annual deviation series provides the basis for the following descriptive measures:

- **Maximum negative deviation:** identifies the lowest observed deviation from the selected reference level or trend.
- **Cumulative signed deviation:** summarizes the direction and magnitude of deviations over a defined observation window.
- **Recovery duration:** may indicate how long it takes for an indicator to return to a defined reference threshold.

The interpretation of these measures depends on the economic meaning of the respective indicator. A negative deviation does not necessarily represent an economic loss for every indicator.

For this reason, production, standardized domestic use and the self-sufficiency rate are initially examined separately. Recovery thresholds must also be interpreted in relation to the respective indicator.

Recovery duration is not calculated until a consistent recovery criterion has been specified for each indicator. In particular, reaching a reference threshold in a single year is not necessarily evidence of sustained recovery.

### 9.7 Overlapping Crisis Events and Structural Changes

The follow-up period of one crisis may overlap with the beginning or continuation of other crisis events. Such overlaps can influence the observed development of agricultural market indicators.

For each event, other historically documented events beginning within the observation window are therefore identified and documented.

Observed deviations are not automatically attributed to the original event when other relevant disturbances occur during the same period.

Potential structural changes in the BLE data, including changes in statistical definitions, collection methods or product coverage, must also be considered. Such changes may affect the comparability of annual observations and should not be mistaken for crisis-related market responses.

### 9.8 Comparability and Robustness Checks

The robustness of the results is assessed by comparing:

- Three-year and five-year pre-crisis reference periods.
- Deviations from the pre-crisis reference level and from the extrapolated pre-crisis trend.
- Individual annual observations and the short-term, medium-term and long-term follow-up periods.

Relative deviations are used to facilitate comparisons between products with different absolute production and usage levels.

For the self-sufficiency rate, changes are additionally considered in percentage points, since percentage-point changes and relative percentage changes describe different aspects of the indicator.

Results are compared only where the required observations are available. Missing or incomplete reference periods and follow-up windows are documented explicitly.

### 9.9 Scope and Limitations

The analysis uses annual BLE agricultural balance data. Consequently, the results describe changes at the agricultural-year level and do not identify exact monthly responses.

The comparison with pre-crisis levels and trends is descriptive. It does not, by itself, establish that a crisis caused an observed change. Other developments, including weather conditions, structural changes, policy interventions and overlapping crisis events, may also have influenced the indicators.

Events without sufficient pre-crisis observations or outside the BLE observation period cannot be evaluated using the same procedure. Such cases are documented rather than assigned artificial reference values.

The results should therefore be interpreted as evidence of observed agricultural market developments in the context of historically documented crisis events, not as isolated causal effects of individual crises.

```python
# Map a historical month to the corresponding BLE agricultural year
def month_to_agricultural_year(month_period):
    if pd.isna(month_period):
        return None

    year = month_period.year
    month = month_period.month

    start_year = year if month >= 7 else year - 1

    return f"{start_year}/{str(start_year + 1)[-2:]}"


# Create a table assigning each crisis event to a BLE agricultural year
event_year_records = []

for event in crisis_events:
    event_year = month_to_agricultural_year(
        pd.Period(event["Start_Month"], freq="M")
    )

    event_year_records.append({
        "Event_ID": event["Event_ID"],
        "Event_Name": event["Event_Name"],
        "Start_Month": event["Start_Month"],
        "Event_Year": event_year,
        "Event_Year_Start": int(event_year[:4])
    })

event_years_df = pd.DataFrame(event_year_records)

display(event_years_df)
```

```python
# Ensure that the relevant variables have the correct data types
analysis_df = analysis_df.copy()

analysis_df["Year"] = pd.to_numeric(
    analysis_df["Year"],
    errors="coerce"
)

for indicator in indicators:
    analysis_df[indicator] = pd.to_numeric(
        analysis_df[indicator],
        errors="coerce"
    )

# Sort observations chronologically within each product
analysis_df = analysis_df.sort_values(
    ["Bilanz_Produkt", "Year"]
).reset_index(drop=True)
```

```python
def calculate_event_trajectory(
    data,
    event,
    product,
    indicator,
    pre_years=3,
    post_years=10
):
    event_year = int(
        month_to_agricultural_year(
            pd.Period(event["Start_Month"], freq="M")
        )[:4]
    )

    product_data = data[
        data["Bilanz_Produkt"] == product
    ].copy()

    # Check whether the event year is within the BLE observation period
    if event_year not in product_data["Year"].values:
        return None, "Event year not available in BLE data"

    # Select the required pre-crisis reference period
    pre_period = product_data[
        (product_data["Year"] >= event_year - pre_years) &
        (product_data["Year"] < event_year)
    ].copy()

    pre_period = pre_period.sort_values("Year")

    # Require a complete pre-crisis reference period
    expected_pre_years = list(
        range(event_year - pre_years, event_year)
    )

    available_pre_years = pre_period["Year"].dropna().astype(int).tolist()

    if available_pre_years != expected_pre_years:
        return None, "Incomplete pre-crisis reference period"

    if pre_period[indicator].isna().any():
        return None, "Missing values in pre-crisis reference period"

    # Calculate the pre-crisis reference level
    pre_crisis_mean = pre_period[indicator].mean()

    # Estimate the linear pre-crisis trend
    trend_coefficients = np.polyfit(
        pre_period["Year"],
        pre_period[indicator],
        deg=1
    )

    # Select the event year and available follow-up observations
    post_period = product_data[
        (product_data["Year"] >= event_year) &
        (product_data["Year"] <= event_year + post_years)
    ].copy()

    post_period = post_period.sort_values("Year")

    if post_period.empty:
        return None, "No event or follow-up observations available"

    # Calculate event time
    post_period["Event_Time"] = (
        post_period["Year"] - event_year
    ).astype(int)

    # Assign analytical phases
    post_period["Phase"] = np.select(
        [
            post_period["Event_Time"] == 0,
            post_period["Event_Time"].between(1, 2),
            post_period["Event_Time"].between(3, 5),
            post_period["Event_Time"].between(6, 10)
        ],
        [
            "Event year",
            "Short-term follow-up",
            "Medium-term follow-up",
            "Long-term follow-up"
        ],
        default="Outside"
    )

    # Calculate the extrapolated pre-crisis trend
    post_period["Pre_Crisis_Trend"] = np.polyval(
        trend_coefficients,
        post_period["Year"]
    )

    # Calculate deviations from the pre-crisis reference level
    post_period["Pre_Crisis_Mean"] = pre_crisis_mean

    post_period["Deviation_From_Mean"] = (
        post_period[indicator] - pre_crisis_mean
    )

    if pre_crisis_mean != 0:
        post_period["Deviation_From_Mean_Pct"] = (
            post_period["Deviation_From_Mean"] /
            abs(pre_crisis_mean)
        ) * 100
    else:
        post_period["Deviation_From_Mean_Pct"] = np.nan

    # Calculate deviations from the extrapolated pre-crisis trend
    post_period["Deviation_From_Trend"] = (
        post_period[indicator] -
        post_period["Pre_Crisis_Trend"]
    )

    post_period["Deviation_From_Trend_Pct"] = np.where(
        post_period["Pre_Crisis_Trend"] != 0,
        (
            post_period["Deviation_From_Trend"] /
            post_period["Pre_Crisis_Trend"].abs()
        ) * 100,
        np.nan
    )

    # Add event and reference-period information
    post_period["Event_ID"] = event["Event_ID"]
    post_period["Event_Name"] = event["Event_Name"]
    post_period["Event_Year"] = event_year
    post_period["Bilanz_Produkt"] = product
    post_period["Indicator"] = indicator
    post_period["Pre_Years"] = pre_years

    return post_period, None
```

```python
# Calculate trajectories using both pre-crisis reference periods
trajectory_results = []
skipped_records = []

for event in crisis_events:

    for product in analysis_df["Bilanz_Produkt"].dropna().unique():

        for indicator in indicators:

            for pre_years in [3, 5]:

                trajectory, skip_reason = calculate_event_trajectory(
                    data=analysis_df,
                    event=event,
                    product=product,
                    indicator=indicator,
                    pre_years=pre_years,
                    post_years=10
                )

                if trajectory is not None:
                    trajectory_results.append(trajectory)

                else:
                    skipped_records.append({
                        "Event_ID": event["Event_ID"],
                        "Event_Name": event["Event_Name"],
                        "Bilanz_Produkt": product,
                        "Indicator": indicator,
                        "Pre_Years": pre_years,
                        "Reason": skip_reason
                    })


# Combine all successful trajectories
if trajectory_results:
    trajectory_df = pd.concat(
        trajectory_results,
        ignore_index=True
    )
else:
    trajectory_df = pd.DataFrame()


# Store all skipped calculations
skipped_df = pd.DataFrame(skipped_records)


# Display calculation results
print("Trajectory calculations completed.")
print("-----------------------------------")
print("Successful calculations:", len(trajectory_results))
print("Skipped calculations:", len(skipped_df))
print("Total annual trajectory observations:", len(trajectory_df))


# Display skipped calculations
if not skipped_df.empty:
    display(skipped_df)
else:
    print("No calculations were skipped.")


# Check the resulting analytical phases
if not trajectory_df.empty:
    print("\nAvailable observations by analytical phase:")
    display(
        trajectory_df.groupby("Phase")["Event_Time"]
        .agg(["count", "min", "max"])
    )
```

Annual deviations are calculated relative to the mean value of the respective indicator during the pre-crisis reference period.

The main specification uses the three agricultural years preceding the event year (\(t=-3\) to \(t=-1\)).

A five-year pre-crisis reference period (\(t=-5\) to \(t=-1\)) is additionally examined as a sensitivity specification.

The percentage deviation is calculated as:

\[
D_t =
\frac{X_t-\overline{X}_{pre}}{\overline{X}_{pre}}
\times 100
\]

where \(X_t\) denotes the observed value in event time \(t\), and \(\overline{X}_{pre}\) represents the mean value during the selected pre-crisis reference period.

For the self-sufficiency rate, the heatmap displays percentage deviations relative to the pre-crisis mean, not absolute percentage-point changes. These two measures must be distinguished when interpreting changes in the self-sufficiency rate.

### 10.1.5 Colour Scale and Missing Observations

The heatmaps use a diverging colour scale centred on zero.

Negative values indicate observations below the pre-crisis reference mean, while positive values indicate observations above that reference mean. Values close to zero indicate relatively small deviations from the reference level.

For each indicator, a common symmetric colour scale is applied across all three product panels. This ensures that the same colour represents the same magnitude and direction of deviation within the respective indicator.

Grey cells indicate unavailable observations. They do not represent zero deviations, an absence of crisis effects, or evidence of market stability.

### 10.1.6 Interpretation of the Heatmaps

The heatmaps provide a descriptive overview of the magnitude, direction, and temporal development of deviations following individual crisis events.

They allow the identification of patterns such as:

- pronounced negative deviations in the event year or subsequent years;
- positive deviations relative to the pre-crisis reference mean;
- persistence of deviations over several agricultural years;
- differences in adjustment patterns between products and indicators;
- variations in the observed trajectories across crisis events.

The heatmaps do not independently establish whether a deviation represents an improvement or deterioration in market resilience. This depends on the economic meaning of the respective indicator.

Furthermore, the event-aligned presentation facilitates temporal comparison but does not establish causal relationships between crisis events and observed market developments.

### 10.1.7 Limitations

The BLE supply balance data are available at annual agricultural-year resolution, covering the period from 1 July to 30 June. Consequently, the heatmaps cannot identify the exact month in which a market indicator responds to a crisis.

The event year \(t=0\) may contain both pre-crisis and post-crisis months. It therefore does not represent a full year of crisis exposure.

The length of the observable post-crisis period also varies across events. Recent events have fewer available follow-up years than earlier events. Grey cells consequently reflect limitations in the observation period rather than necessarily indicating an absence of subsequent market adjustment.

Finally, deviations in later agricultural years may coincide with subsequent crisis events or structural changes in agricultural markets. Long-term patterns should therefore not be attributed exclusively to the initial crisis event.

The heatmaps are thus used as an exploratory and comparative visualization of observed market adjustment patterns. Their findings are complemented by the event-specific trajectory analysis and the resilience indicators presented in the following sections.

## 10.3 Calculation and Visualization of Resilience Indicators

### 10.3.1 Objective

This section operationalizes market resilience through a set of descriptive indicators derived from the annual deviations of agricultural market variables from their respective pre-crisis reference levels. The objective is to assess the magnitude of negative deviations, the extent to which indicators return to their pre-crisis reference range, and the cumulative and average deviations observed during the post-crisis period.

The analysis is conducted separately for barley, potatoes, and soft wheat and considers the three indicators of usable production, standardized domestic use, and the self-sufficiency rate.

The calculations are based on the annual trajectories generated in Step 9, using the three-year pre-crisis reference period as the main specification.

### 10.3.2 Deepest Negative Deviation

The deepest negative deviation measures the largest observed percentage decline below the pre-crisis reference mean during the available post-crisis observation period.

It is calculated as the minimum negative annual deviation from the pre-crisis mean:

\[
D_{\min}=\min_{t \in T}(D_t), \quad D_t<0
\]

where \(D_t\) represents the percentage deviation from the pre-crisis reference mean in event year \(t\), and \(T\) denotes the available observation period from \(t=0\) to the final observed follow-up year.

If no negative deviation is observed, the indicator is recorded as unavailable rather than assigning a value of zero. This distinction prevents the absence of a negative deviation from being interpreted as an observed negative deviation of zero.

The event time at which the deepest negative deviation occurs is additionally recorded as the trough event time.

### 10.3.3 Recovery Duration

Recovery is operationally defined as the return of an indicator to a corridor of ±5% around its pre-crisis reference mean for at least two consecutive observed agricultural years.

The recovery criterion is therefore:

\[
|D_t| \leq 5\%
\]

for two consecutive observed years.

Recovery duration is measured as the number of years between the event time of the deepest negative deviation and the beginning of the first subsequent two-year period satisfying this criterion:

\[
R=t_{\text{recovery}}-t_{\text{trough}}
\]

where \(t_{\text{trough}}\) is the event time of the deepest negative deviation and \(t_{\text{recovery}}\) is the beginning of the first qualifying recovery period.

The analysis distinguishes three outcomes:

- **Recovered:** The recovery criterion is met within the available observation period.
- **Not observed within available follow-up:** A negative deviation occurred, but the recovery criterion was not met within the observed period.
- **No negative deviation observed:** No negative deviation from the pre-crisis reference mean occurred during the available observation period.

The recovery threshold and the requirement of two consecutive years constitute an operational definition. They are applied consistently across products and indicators to facilitate comparison.

### 10.3.4 Cumulative Deviation

Cumulative deviation measures the sum of the annual percentage deviations from the pre-crisis reference mean over the available observation period:

\[
C=\sum_{t \in T}D_t
\]

Positive and negative deviations are retained with their respective signs.

This measure captures the cumulative direction and magnitude of deviations over time. It is not interpreted as a direct measure of economic loss or welfare change. Its meaning depends on the indicator considered: for example, an increase in domestic use does not necessarily represent an improvement or deterioration in market resilience.

Since the number of available follow-up years differs between crisis events, cumulative deviations are interpreted with regard to the respective observation period.

### 10.3.5 Mean Annual Deviation

To complement cumulative deviation, mean annual deviation is calculated as the arithmetic mean of the observed annual percentage deviations:

\[
\overline{D}=\frac{1}{n}\sum_{t \in T}D_t
\]

where \(n\) is the number of available annual observations in the respective event trajectory.

This measure provides the average annual deviation from the pre-crisis reference mean and reduces the direct dependence of the cumulative measure on the length of the observation period.

However, it does not eliminate differences in the composition of observation periods, subsequent crisis exposure, or structural changes. Comparisons between events with different follow-up lengths therefore remain subject to these limitations.

### 10.3.6 Observation Coverage and Comparability

The maximum follow-up period is ten years after the event year (\(t=0\)). However, the available observation period varies between crisis events because the BLE supply balance data cover the agricultural years 1991/92 to 2024/25.

Consequently, earlier events have longer observable post-crisis trajectories than more recent events. The number of available observations is therefore reported alongside the resilience indicators.

The absence of an observed recovery within a limited follow-up period must not be interpreted as evidence that recovery did not occur subsequently. It indicates only that recovery could not be established within the available data.

Furthermore, long-term deviations may overlap with subsequent crisis events and structural changes in agricultural markets. They cannot therefore be attributed exclusively to the initially examined crisis.

### 10.3.7 Interpretation and Limitations

The resilience indicators provide complementary descriptive perspectives on agricultural market developments following crisis events.

The deepest negative deviation captures the magnitude of the largest observed negative departure from the pre-crisis reference mean. Recovery duration describes the time required to return to the defined reference corridor after the deepest negative deviation. Cumulative and mean annual deviations summarize the direction and magnitude of deviations over the available observation period.

The indicators are interpreted separately for each product and market variable. In particular, changes in usable production, domestic use, and the self-sufficiency rate do not have identical economic implications. A positive deviation must therefore not automatically be interpreted as an improvement in resilience, nor a negative deviation as an equivalent deterioration across all indicators.

Finally, the analysis is descriptive and does not establish causal relationships between crisis events and observed market developments. The annual resolution of the BLE data also limits the identification of the exact timing of within-year adjustments. The results should consequently be interpreted as indicators of observed market adjustment and resilience patterns rather than causal estimates of crisis impacts.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ==================================================
# STEP 10.3 – RESILIENCE INDICATORS
# ==================================================


# --------------------------------------------------
# 1. Configuration
# --------------------------------------------------

products = ["Weichweizen", "Gerste", "Kartoffeln"]

product_labels = {
    "Weichweizen": "Soft wheat",
    "Gerste": "Barley",
    "Kartoffeln": "Potatoes"
}

indicator_labels = {
    "Verwendbare_Erzeugung": "Usable production",
    "Inlandsverwendung_standardisiert": "Standardized domestic use",
    "Selbstversorgungsgrad_in_%": "Self-sufficiency rate"
}

# Main specification
pre_years = 3

# Recovery definition
recovery_threshold = 5.0
required_recovery_years = 2


# --------------------------------------------------
# 2. Prepare trajectory data
# --------------------------------------------------

resilience_data = trajectory_df[
    trajectory_df["Pre_Years"] == pre_years
].copy()

resilience_data["Event_Time"] = pd.to_numeric(
    resilience_data["Event_Time"],
    errors="coerce"
)

resilience_data["Deviation_From_Mean_Pct"] = pd.to_numeric(
    resilience_data["Deviation_From_Mean_Pct"],
    errors="coerce"
)

if resilience_data.empty:
    raise ValueError(
        "No trajectory data available for the selected reference period."
    )


# --------------------------------------------------
# 3. Calculate resilience metrics
# --------------------------------------------------

def calculate_resilience_metrics(group):

    group = (
        group
        .dropna(
            subset=[
                "Event_Time",
                "Deviation_From_Mean_Pct"
            ]
        )
        .sort_values("Event_Time")
        .copy()
    )

    values = group[
        "Deviation_From_Mean_Pct"
    ].to_numpy(dtype=float)

    event_times = group[
        "Event_Time"
    ].to_numpy(dtype=int)

    available_years = len(values)

    if available_years == 0:

        return {
            "Deepest_Negative_Deviation_Pct": np.nan,
            "Trough_Event_Time": np.nan,
            "Recovery_Start_Event_Time": np.nan,
            "Recovery_Duration_Years": np.nan,
            "Cumulative_Deviation_Pct": np.nan,
            "Mean_Annual_Deviation_Pct": np.nan,
            "Available_Years": 0,
            "Recovery_Status": "No observations"
        }

    # 3.1 Deepest negative deviation and trough

    negative_indices = np.where(values < 0)[0]

    if len(negative_indices) > 0:

        trough_index = negative_indices[
            np.argmin(values[negative_indices])
        ]

        deepest_negative = values[trough_index]
        trough_event_time = event_times[trough_index]

    else:

        deepest_negative = np.nan
        trough_event_time = np.nan

    # 3.2 Recovery after the deepest negative deviation

    recovery_start_event_time = np.nan
    recovery_duration_years = np.nan

    if len(negative_indices) == 0:

        recovery_status = "No negative deviation observed"

    else:

        recovery_status = (
            "Not observed within available follow-up"
        )

        # Search only after the deepest negative deviation.
        # Recovery requires two consecutive observed years
        # within the +/- 5% deviation corridor.

        for i in range(
            len(values) - required_recovery_years + 1
        ):

            current_times = event_times[
                i:i + required_recovery_years
            ]

            current_values = values[
                i:i + required_recovery_years
            ]

            consecutive_years = np.all(
                np.diff(current_times) == 1
            )

            after_trough = (
                current_times[0] > trough_event_time
            )

            within_recovery_corridor = np.all(
                np.abs(current_values) <= recovery_threshold
            )

            if (
                consecutive_years
                and after_trough
                and within_recovery_corridor
            ):

                recovery_start_event_time = current_times[0]

                recovery_duration_years = (
                    recovery_start_event_time - trough_event_time
                )

                recovery_status = "Recovered"
                break

    # 3.3 Cumulative signed deviation

    cumulative_deviation = values.sum()

    # 3.4 Mean annual deviation

    mean_annual_deviation = (
        cumulative_deviation / available_years
    )

    return {
        "Deepest_Negative_Deviation_Pct": deepest_negative,
        "Trough_Event_Time": trough_event_time,
        "Recovery_Start_Event_Time": recovery_start_event_time,
        "Recovery_Duration_Years": recovery_duration_years,
        "Cumulative_Deviation_Pct": cumulative_deviation,
        "Mean_Annual_Deviation_Pct": mean_annual_deviation,
        "Available_Years": available_years,
        "Recovery_Status": recovery_status
    }


# --------------------------------------------------
# 4. Calculate metrics for each event, product
#    and indicator
# --------------------------------------------------

group_columns = [
    "Event_ID",
    "Bilanz_Produkt",
    "Indicator"
]

resilience_results = []

for group_values, group in resilience_data.groupby(
    group_columns
):

    metrics = calculate_resilience_metrics(group)

    result = dict(zip(group_columns, group_values))
    result.update(metrics)

    resilience_results.append(result)


resilience_metrics = pd.DataFrame(
    resilience_results
)


# --------------------------------------------------
# 5. Organize and sort results
# --------------------------------------------------

resilience_metrics = resilience_metrics[
    [
        "Event_ID",
        "Bilanz_Produkt",
        "Indicator",
        "Deepest_Negative_Deviation_Pct",
        "Trough_Event_Time",
        "Recovery_Start_Event_Time",
        "Recovery_Duration_Years",
        "Cumulative_Deviation_Pct",
        "Mean_Annual_Deviation_Pct",
        "Available_Years",
        "Recovery_Status"
    ]
]

resilience_metrics = (
    resilience_metrics
    .sort_values(
        ["Event_ID", "Indicator", "Bilanz_Produkt"]
    )
    .reset_index(drop=True)
)


# --------------------------------------------------
# 6. Display results
# --------------------------------------------------

display(resilience_metrics)


# --------------------------------------------------
# 7. Data quality and coverage checks
# --------------------------------------------------

print("Number of resilience calculations:")
print(len(resilience_metrics))

print("\nEvents included:")
print(sorted(resilience_metrics["Event_ID"].unique()))

print("\nProduct coverage:")
print(
    resilience_metrics["Bilanz_Produkt"].value_counts()
)

print("\nIndicator coverage:")
print(
    resilience_metrics["Indicator"].value_counts()
)

print("\nRecovery status:")
print(
    resilience_metrics["Recovery_Status"]
    .value_counts(dropna=False)
)

print("\nAvailable observation years by event:")

display(
    resilience_metrics
    .groupby("Event_ID")["Available_Years"]
    .agg(["min", "max"])
)


# --------------------------------------------------
# 8. Plot 1 – Deepest negative deviation
# --------------------------------------------------

for indicator in indicators:

    plot_data = resilience_metrics[
        resilience_metrics["Indicator"] == indicator
    ].copy()

    fig, ax = plt.subplots(figsize=(12, 6))

    for product in products:

        product_data = plot_data[
            plot_data["Bilanz_Produkt"] == product
        ].sort_values("Event_ID")

        ax.plot(
            product_data["Event_ID"],
            product_data["Deepest_Negative_Deviation_Pct"],
            marker="o",
            linewidth=2,
            label=product_labels[product]
        )

    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1
    )

    ax.set_title(
        f"Deepest Negative Deviation – "
        f"{indicator_labels[indicator]}",
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel("Crisis event")
    ax.set_ylabel(
        "Deepest deviation from pre-crisis mean (%)"
    )

    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend(title="Product")

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# 9. Plot 2 – Recovery duration
# --------------------------------------------------

for indicator in indicators:

    plot_data = resilience_metrics[
        resilience_metrics["Indicator"] == indicator
    ].copy()

    fig, ax = plt.subplots(figsize=(12, 6))

    for product in products:

        product_data = plot_data[
            plot_data["Bilanz_Produkt"] == product
        ].sort_values("Event_ID")

        ax.plot(
            product_data["Event_ID"],
            product_data["Recovery_Duration_Years"],
            marker="o",
            linewidth=2,
            label=product_labels[product]
        )

    ax.set_title(
        f"Recovery Duration – "
        f"{indicator_labels[indicator]}",
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel("Crisis event")
    ax.set_ylabel(
        "Years from trough to sustained recovery"
    )

    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend(title="Product")

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# 10. Plot 3 – Cumulative deviation
# --------------------------------------------------

for indicator in indicators:

    plot_data = resilience_metrics[
        resilience_metrics["Indicator"] == indicator
    ].copy()

    fig, ax = plt.subplots(figsize=(12, 6))

    for product in products:

        product_data = plot_data[
            plot_data["Bilanz_Produkt"] == product
        ].sort_values("Event_ID")

        ax.plot(
            product_data["Event_ID"],
            product_data["Cumulative_Deviation_Pct"],
            marker="o",
            linewidth=2,
            label=product_labels[product]
        )

    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1
    )

    ax.set_title(
        f"Cumulative Deviation – "
        f"{indicator_labels[indicator]}",
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel("Crisis event")
    ax.set_ylabel(
        "Sum of annual percentage deviations"
    )

    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend(title="Product")

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# 11. Plot 4 – Mean annual deviation
# --------------------------------------------------

for indicator in indicators:

    plot_data = resilience_metrics[
        resilience_metrics["Indicator"] == indicator
    ].copy()

    fig, ax = plt.subplots(figsize=(12, 6))

    for product in products:

        product_data = plot_data[
            plot_data["Bilanz_Produkt"] == product
        ].sort_values("Event_ID")

        ax.plot(
            product_data["Event_ID"],
            product_data["Mean_Annual_Deviation_Pct"],
            marker="o",
            linewidth=2,
            label=product_labels[product]
        )

    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1
    )

    ax.set_title(
        f"Mean Annual Deviation – "
        f"{indicator_labels[indicator]}",
        fontsize=13,
        fontweight="bold"
    )

    ax.set_xlabel("Crisis event")
    ax.set_ylabel(
        "Mean annual deviation from pre-crisis mean (%)"
    )

    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend(title="Product")

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# 12. Export results
# --------------------------------------------------

output_file = "crisis_resilience_metrics.xlsx"

resilience_metrics.to_excel(
    output_file,
    index=False,
    engine="openpyxl"
)

print(
    f"\nResilience metrics exported to: {output_file}"
)
```

## 10.4 Sensitivity Analysis of Resilience Indicators

### 10.4.1 Objective

The sensitivity analysis examines whether the assessment of agricultural market resilience depends on the length of the pre-crisis reference period. The main analysis uses a three-year pre-crisis reference period, while a five-year reference period is applied as a sensitivity specification.

Both specifications use the same crisis events, products, indicators, post-crisis observations, and definitions of the resilience metrics. This allows differences in the results to be attributed to the choice of reference period rather than to differences in the observation window.

### 10.4.2 Comparison of Reference Periods

For each event, product, and indicator, the resilience metrics calculated using the three-year and five-year reference periods are compared.

The comparison includes:

- Deepest negative deviation from the pre-crisis reference level
- Recovery duration
- Cumulative deviation
- Mean annual deviation
- Recovery status

Differences are calculated as:

\[
\Delta M = M_{5Y} - M_{3Y}
\]

where \(M_{5Y}\) denotes the metric calculated using the five-year reference period and \(M_{3Y}\) the corresponding metric based on the three-year reference period.

A positive or negative difference indicates that the metric changes when the reference period is extended. The direction of the difference does not necessarily indicate an improvement or deterioration in resilience; interpretation depends on the respective indicator and metric.

### 10.4.3 Matched Comparisons and Data Coverage

The paired sensitivity analysis includes only cases for which both reference-period specifications are available.

A total of 72 matched comparisons were obtained, covering eight crisis events (EV04–EV11), three agricultural products, and three indicators.

EV03 is excluded from the paired comparison because the available data do not provide a complete five-year pre-crisis reference period. It remains included in the main analysis based on the three-year reference period.

The number of available post-crisis observations is identical under both specifications for every included event. However, the length of the post-crisis observation period differs between events, ranging from two observations for EV11 to eleven observations for EV04 and EV05. This limits the comparability of recovery-related metrics across events, particularly for more recent crises.

### 10.4.4 Sensitivity of Recovery Classification

The recovery status is identical under both reference-period specifications in 62 of the 72 matched comparisons. In the remaining ten cases, the classification differs.

Agreement between the two specifications indicates that the recovery classification is not sensitive to the choice of reference period in those cases. However, agreement does not necessarily imply that recovery was observed: both specifications may classify recovery as not observed within the available follow-up period.

Differences in recovery classification are examined separately to identify the events, products, and indicators for which the choice of reference period affects the assessment.

Recovery duration is interpreted only where the defined recovery criterion is met. Cases without an observed recovery within the available follow-up period are not assigned an artificial recovery duration.

### 10.4.5 Interpretation and Limitations

The sensitivity analysis assesses the robustness of the descriptive resilience measures to the choice of pre-crisis reference period. It does not establish causal effects of crisis events.

The three-year reference period remains the main specification, while the five-year reference period provides a robustness check. The results indicate that the choice of reference period can affect individual metric values and, in some cases, the recovery classification.

The limited post-crisis observation periods of recent events must be considered when interpreting recovery-related results. In particular, the absence of an observed recovery does not demonstrate that a market will not recover; it means only that recovery was not identified within the available observation period.

Overall, the sensitivity analysis provides an indication of the extent to which the assessment of agricultural market resilience depends on the selected pre-crisis reference period.

```python

# ============================================================
# STEP 10.4 – Sensitivity Analysis of Resilience Indicators
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

reference_periods = [3, 5]

recovery_threshold = 5.0
required_recovery_years = 2

product_order = ["Weichweizen", "Gerste", "Kartoffeln"]

product_labels = {
    "Weichweizen": "Soft wheat",
    "Gerste": "Barley",
    "Kartoffeln": "Potatoes"
}

indicator_labels = {
    "Verwendbare_Erzeugung": "Usable production",
    "Inlandsverwendung_standardisiert": "Standardised domestic use",
    "Selbstversorgungsgrad_in_%": "Self-sufficiency rate"
}

# ------------------------------------------------------------
# 2. Calculate resilience metrics for each specification
# ------------------------------------------------------------

def calculate_sensitivity_metrics(group):

    group = (
        group
        .dropna(subset=["Event_Time", "Deviation_From_Mean_Pct"])
        .sort_values("Event_Time")
        .copy()
    )

    if group.empty:
        return pd.Series({
            "Deepest_Negative_Deviation_Pct": np.nan,
            "Trough_Event_Time": np.nan,
            "Recovery_Start_Event_Time": np.nan,
            "Recovery_Duration_Years": np.nan,
            "Cumulative_Deviation_Pct": np.nan,
            "Mean_Annual_Deviation_Pct": np.nan,
            "Available_Years": 0,
            "Recovery_Status": "No observations"
        })

    deviations = group["Deviation_From_Mean_Pct"].to_numpy()
    event_times = group["Event_Time"].to_numpy()

    # Deepest negative deviation
    negative_mask = deviations < 0

    if negative_mask.any():

        negative_positions = np.where(negative_mask)[0]
        trough_position = negative_positions[
            np.argmin(deviations[negative_mask])
        ]

        deepest_negative = deviations[trough_position]
        trough_event_time = event_times[trough_position]

    else:
        deepest_negative = np.nan
        trough_event_time = np.nan
        trough_position = None

    # Recovery: two consecutive observed years within the
    # +/- threshold after the deepest negative deviation.
    recovery_start = np.nan
    recovery_duration = np.nan

    if trough_position is not None:

        for i in range(
            trough_position + 1,
            len(group) - required_recovery_years + 2
        ):

            candidate_times = event_times[
                i:i + required_recovery_years
            ]

            candidate_values = deviations[
                i:i + required_recovery_years
            ]

            # Require consecutive event years
            consecutive_years = np.all(
                np.diff(candidate_times) == 1
            )

            within_threshold = np.all(
                np.abs(candidate_values) <= recovery_threshold
            )

            if consecutive_years and within_threshold:

                recovery_start = candidate_times[0]

                recovery_duration = (
                    recovery_start - trough_event_time
                )

                break

    # Recovery status
    if trough_position is None:

        recovery_status = "No negative deviation observed"

    elif pd.notna(recovery_start):

        recovery_status = "Recovered"

    else:

        recovery_status = (
            "Not observed within available follow-up"
        )

    # Cumulative and mean annual deviation
    cumulative_deviation = deviations.sum()
    mean_annual_deviation = deviations.mean()

    return pd.Series({
        "Deepest_Negative_Deviation_Pct": deepest_negative,
        "Trough_Event_Time": trough_event_time,
        "Recovery_Start_Event_Time": recovery_start,
        "Recovery_Duration_Years": recovery_duration,
        "Cumulative_Deviation_Pct": cumulative_deviation,
        "Mean_Annual_Deviation_Pct": mean_annual_deviation,
        "Available_Years": len(group),
        "Recovery_Status": recovery_status
    })


# ------------------------------------------------------------
# 3. Calculate metrics for 3-year and 5-year references
# ------------------------------------------------------------

sensitivity_results = []

for pre_years in reference_periods:

    subset = trajectory_df[
        trajectory_df["Pre_Years"] == pre_years
    ].copy()

    grouped = subset.groupby(
        ["Event_ID", "Bilanz_Produkt", "Indicator"],
        sort=False
    )

    for (event_id, product, indicator), group in grouped:

        metrics = calculate_sensitivity_metrics(group)

        result = {
            "Event_ID": event_id,
            "Bilanz_Produkt": product,
            "Indicator": indicator,
            "Pre_Years": pre_years
        }

        result.update(metrics.to_dict())

        sensitivity_results.append(result)


sensitivity_metrics = pd.DataFrame(sensitivity_results)

print("Sensitivity metrics calculated:")
print(sensitivity_metrics.shape)

# ------------------------------------------------------------
# 4. Create separate tables for 3-year and 5-year references
# ------------------------------------------------------------

metric_columns = [
    "Deepest_Negative_Deviation_Pct",
    "Trough_Event_Time",
    "Recovery_Start_Event_Time",
    "Recovery_Duration_Years",
    "Cumulative_Deviation_Pct",
    "Mean_Annual_Deviation_Pct",
    "Available_Years",
    "Recovery_Status"
]

results_3y = (
    sensitivity_metrics[
        sensitivity_metrics["Pre_Years"] == 3
    ]
    .drop(columns="Pre_Years")
    .rename(columns={
        column: f"{column}_3Y"
        for column in metric_columns
    })
)

results_5y = (
    sensitivity_metrics[
        sensitivity_metrics["Pre_Years"] == 5
    ]
    .drop(columns="Pre_Years")
    .rename(columns={
        column: f"{column}_5Y"
        for column in metric_columns
    })
)

# ------------------------------------------------------------
# 5. Match both specifications
# ------------------------------------------------------------

comparison_keys = [
    "Event_ID",
    "Bilanz_Produkt",
    "Indicator"
]

sensitivity_comparison = pd.merge(
    results_3y,
    results_5y,
    on=comparison_keys,
    how="inner",
    validate="one_to_one"
)

# Calculate differences: 5-year minus 3-year reference
numeric_metrics = [
    "Deepest_Negative_Deviation_Pct",
    "Recovery_Duration_Years",
    "Cumulative_Deviation_Pct",
    "Mean_Annual_Deviation_Pct"
]

for metric in numeric_metrics:

    sensitivity_comparison[
        f"{metric}_Difference_5Y_minus_3Y"
    ] = (
        sensitivity_comparison[f"{metric}_5Y"]
        - sensitivity_comparison[f"{metric}_3Y"]
    )

# Compare recovery classifications
sensitivity_comparison["Recovery_Status_Consistent"] = (
    sensitivity_comparison["Recovery_Status_3Y"]
    == sensitivity_comparison["Recovery_Status_5Y"]
)

# ------------------------------------------------------------
# 6. Identify cases with different recovery classifications
# ------------------------------------------------------------

recovery_status_differences = (
    sensitivity_comparison[
        ~sensitivity_comparison["Recovery_Status_Consistent"]
    ]
    .copy()
    .sort_values(comparison_keys)
)

# ------------------------------------------------------------
# 7. Data coverage checks
# ------------------------------------------------------------

print("\nNumber of matched comparisons:")
print(len(sensitivity_comparison))

print("\nEvents included in paired comparison:")
print(sorted(sensitivity_comparison["Event_ID"].unique()))

print("\nProduct coverage:")
print(
    sensitivity_comparison["Bilanz_Produkt"]
    .value_counts()
    .sort_index()
)

print("\nIndicator coverage:")
print(
    sensitivity_comparison["Indicator"]
    .value_counts()
    .sort_index()
)

print("\nRecovery status agreement:")
print(
    sensitivity_comparison["Recovery_Status_Consistent"]
    .value_counts()
)

print("\nCases with different recovery classifications:")
display(recovery_status_differences)

# Check that post-crisis observation counts are identical
coverage_check = sensitivity_comparison[
    [
        "Event_ID",
        "Available_Years_3Y",
        "Available_Years_5Y"
    ]
].drop_duplicates().sort_values("Event_ID")

print("\nAvailable follow-up years under both specifications:")
display(coverage_check)

# ------------------------------------------------------------
# 8. Sensitivity plots
# ------------------------------------------------------------

plot_metrics = [
    (
        "Deepest_Negative_Deviation_Pct_Difference_5Y_minus_3Y",
        "Difference in deepest negative deviation (5Y - 3Y)"
    ),
    (
        "Recovery_Duration_Years_Difference_5Y_minus_3Y",
        "Difference in recovery duration (5Y - 3Y)"
    ),
    (
        "Cumulative_Deviation_Pct_Difference_5Y_minus_3Y",
        "Difference in cumulative deviation (5Y - 3Y)"
    ),
    (
        "Mean_Annual_Deviation_Pct_Difference_5Y_minus_3Y",
        "Difference in mean annual deviation (5Y - 3Y)"
    )
]

for indicator in sensitivity_comparison["Indicator"].unique():

    indicator_data = sensitivity_comparison[
        sensitivity_comparison["Indicator"] == indicator
    ].copy()

    for metric, metric_label in plot_metrics:

        fig, ax = plt.subplots(figsize=(12, 6))

        for product in product_order:

            product_data = indicator_data[
                indicator_data["Bilanz_Produkt"] == product
            ].sort_values("Event_ID")

            if product_data.empty:
                continue

            ax.plot(
                product_data["Event_ID"],
                product_data[metric],
                marker="o",
                label=product_labels.get(product, product)
            )

        ax.axhline(
            y=0,
            linestyle="--",
            linewidth=1
        )

        ax.set_title(
            f"{metric_label}\n"
            f"{indicator_labels.get(indicator, indicator)}"
        )

        ax.set_xlabel("Crisis event")
        ax.set_ylabel("Difference (5-year minus 3-year reference)")
        ax.legend(title="Agricultural product")
        ax.grid(axis="y", alpha=0.3)

        plt.tight_layout()
        plt.show()

# ------------------------------------------------------------
# 9. Export results
# ------------------------------------------------------------

output_path = Path(
    "crisis_resilience_sensitivity_analysis.xlsx"
)

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:

    sensitivity_metrics.to_excel(
        writer,
        sheet_name="All_Specifications",
        index=False
    )

    sensitivity_comparison.to_excel(
        writer,
        sheet_name="Paired_Comparison",
        index=False
    )

    recovery_status_differences.to_excel(
        writer,
        sheet_name="Recovery_Differences",
        index=False
    )

    coverage_check.to_excel(
        writer,
        sheet_name="Coverage_Check",
        index=False
    )

print("\nSensitivity analysis completed.")
print(f"Matched comparisons: {len(sensitivity_comparison)}")
print(
    "Different recovery classifications:",
    len(recovery_status_differences)
)
print(f"Results exported to: {output_path.resolve()}")
```

### 10.4.6 Summary of Sensitivity Analysis

The sensitivity analysis compared resilience metrics based on three-year and five-year pre-crisis reference periods. A total of 72 matched comparisons were obtained, covering eight crisis events (EV04–EV11), three agricultural products, and three indicators. EV03 was excluded from the paired comparison because a complete five-year pre-crisis reference period was not available.

The post-crisis observation windows were identical under both reference specifications for all matched events. However, the available follow-up periods differed between events, ranging from two years for EV11 to eleven years for EV04 and EV05.

The recovery classification was consistent between the two specifications in 62 of the 72 comparisons (86.1%). In ten cases (13.9%), the recovery status differed depending on the selected reference period. These differences demonstrate that the assessment of recovery is sensitive to the definition of the pre-crisis reference level in some event-product-indicator combinations.

The differing classifications were observed in the following cases:

- EV05: standardised domestic use for soft wheat
- EV06: usable production for potatoes
- EV07: self-sufficiency rate and usable production for barley, and usable production for potatoes
- EV08, EV09 and EV10: usable production for potatoes
- EV11: standardised domestic use and self-sufficiency rate for barley

The repeated findings for potato production in EV08, EV09 and EV10 must be interpreted in light of overlapping crisis periods and shared post-crisis observations. They should not be treated as three independent confirmations of the same effect.

The results also highlight the importance of distinguishing between the recovery classifications. *No negative deviation observed* indicates that no negative deviation from the respective pre-crisis reference level occurred during the observed period. *Not observed within available follow-up* indicates that a negative deviation occurred, but the defined recovery criterion was not met within the available observation period.

Overall, the sensitivity analysis supports the use of the three-year pre-crisis reference period as the main specification, while the five-year reference period provides an additional robustness check. The majority of recovery classifications remain unchanged, although individual results are sensitive to the choice of reference period.

The findings remain descriptive. They do not establish causal effects of individual crisis events. Furthermore, the limited follow-up periods of recent events, particularly EV11, restrict conclusions about longer-term recovery. The sensitivity analysis therefore provides a robustness assessment of the chosen reference period rather than definitive evidence of market resilience.

```python

# ============================================================
# STEP 11.1 – Consolidated Results
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

comparison_keys = [
    "Event_ID",
    "Bilanz_Produkt",
    "Indicator"
]

# ------------------------------------------------------------
# 2. Create historical overview of all 12 crisis events
# ------------------------------------------------------------

event_overview_records = []

# Identify agricultural years available in the BLE dataset
ble_years = set(analysis_df["Wirtschaftsjahr"].dropna().astype(str))

# Main specification: three-year pre-crisis reference
main_metrics = sensitivity_metrics[
    sensitivity_metrics["Pre_Years"] == 3
].copy()

# Events with empirical results under the main specification
events_with_main_results = set(
    main_metrics["Event_ID"].unique()
)

# Events with an event year represented in the BLE data
events_with_ble_event_year = set(
    trajectory_df.loc[
        trajectory_df["Event_Time"] == 0,
        "Event_ID"
    ].unique()
)

for event in crisis_events:

    event_record = event.copy()

    event_id = event["Event_ID"]
    start_month = event["Start_Month"]

    # Map historical start month to BLE agricultural year
    event_agricultural_year = month_to_agricultural_year(
        pd.Period(start_month, freq="M")
    )

    event_record["Event_Agricultural_Year"] = (
        event_agricultural_year
    )

    # Check whether the event year is covered by the BLE dataset
    event_record["BLE_Event_Year_Available"] = (
        event_agricultural_year in ble_years
    )

    # Count empirical product-indicator combinations
    event_record["Main_Analysis_Combinations"] = int(
        (
            main_metrics["Event_ID"] == event_id
        ).sum()
    )

    if event_id in events_with_main_results:

        event_record["Empirical_Status"] = (
            "Included in main analysis (3-year reference)"
        )

    elif event_agricultural_year not in ble_years:

        event_record["Empirical_Status"] = (
            "Outside BLE observation period"
        )

    else:

        event_record["Empirical_Status"] = (
            "Insufficient pre-crisis reference period"
        )

    event_overview_records.append(event_record)


event_overview_df = pd.DataFrame(event_overview_records)

# Preserve chronological order
event_overview_df = event_overview_df.sort_values(
    "Start_Month"
).reset_index(drop=True)

print("Historical event overview:")
print("Number of historical events:", len(event_overview_df))

display(event_overview_df)


# ------------------------------------------------------------
# 3. Create consolidated empirical results
# ------------------------------------------------------------

# Use the three-year reference as the main specification
main_metrics = main_metrics.drop(
    columns=["Pre_Years"]
)

# Keep the five-year sensitivity results separate
sensitivity_5y = (
    sensitivity_metrics[
        sensitivity_metrics["Pre_Years"] == 5
    ]
    .drop(columns=["Pre_Years"])
    .rename(columns={
        column: f"{column}_5Y"
        for column in [
            "Deepest_Negative_Deviation_Pct",
            "Trough_Event_Time",
            "Recovery_Start_Event_Time",
            "Recovery_Duration_Years",
            "Cumulative_Deviation_Pct",
            "Mean_Annual_Deviation_Pct",
            "Available_Years",
            "Recovery_Status"
        ]
    })
)

# Rename main specification columns
main_metrics = main_metrics.rename(columns={
    column: f"{column}_3Y"
    for column in [
        "Deepest_Negative_Deviation_Pct",
        "Trough_Event_Time",
        "Recovery_Start_Event_Time",
        "Recovery_Duration_Years",
        "Cumulative_Deviation_Pct",
        "Mean_Annual_Deviation_Pct",
        "Available_Years",
        "Recovery_Status"
    ]
})

# Merge the main specification with the five-year results
consolidated_results_df = pd.merge(
    main_metrics,
    sensitivity_5y,
    on=comparison_keys,
    how="left",
    validate="one_to_one"
)

# Add the paired sensitivity comparison differences
difference_columns = [
    column
    for column in sensitivity_comparison.columns
    if (
        column.endswith("_Difference_5Y_minus_3Y")
        or column == "Recovery_Status_Consistent"
    )
]

consolidated_results_df = pd.merge(
    consolidated_results_df,
    sensitivity_comparison[
        comparison_keys + difference_columns
    ],
    on=comparison_keys,
    how="left",
    validate="one_to_one"
)

# Add historical event information
historical_columns = [
    "Event_ID",
    "Event_Name",
    "Start_Month",
    "End_Month",
    "Event_Agricultural_Year",
    "Dating_Method",
    "Dating_Status",
    "End_Status",
    "Empirical_Status"
]

consolidated_results_df = pd.merge(
    consolidated_results_df,
    event_overview_df[historical_columns],
    on="Event_ID",
    how="left",
    validate="many_to_one"
)

# Sort results
consolidated_results_df = (
    consolidated_results_df
    .sort_values(
        ["Start_Month", "Bilanz_Produkt", "Indicator"]
    )
    .reset_index(drop=True)
)

print("\nConsolidated empirical results:")
print(consolidated_results_df.shape)

display(consolidated_results_df.head(15))


# ------------------------------------------------------------
# 4. Summarise empirical coverage by crisis event
# ------------------------------------------------------------

trajectory_coverage = (
    trajectory_df[
        trajectory_df["Pre_Years"] == 3
    ]
    .groupby("Event_ID")
    .agg(
        Product_Indicator_Combinations=(
            "Bilanz_Produkt",
            "size"
        ),
        Annual_Trajectory_Observations=(
            "Event_Time",
            "count"
        ),
        First_Event_Time=("Event_Time", "min"),
        Last_Event_Time=("Event_Time", "max")
    )
    .reset_index()
)

event_overview_df = pd.merge(
    event_overview_df,
    trajectory_coverage,
    on="Event_ID",
    how="left"
)

# Events without empirical results remain explicitly missing
coverage_columns = [
    "Product_Indicator_Combinations",
    "Annual_Trajectory_Observations",
    "First_Event_Time",
    "Last_Event_Time"
]

print("\nEmpirical coverage by event:")
display(event_overview_df)


# ------------------------------------------------------------
# 5. Create a compact summary by product and indicator
# ------------------------------------------------------------

cross_event_summary_df = (
    consolidated_results_df
    .groupby(
        ["Bilanz_Produkt", "Indicator"],
        dropna=False
    )
    .agg(
        Number_of_Events=("Event_ID", "nunique"),
        Mean_Annual_Deviation_3Y=(
            "Mean_Annual_Deviation_Pct_3Y",
            "mean"
        ),
        Mean_Deepest_Negative_Deviation_3Y=(
            "Deepest_Negative_Deviation_Pct_3Y",
            "mean"
        ),
        Recovered_Cases_3Y=(
            "Recovery_Status_3Y",
            lambda x: (x == "Recovered").sum()
        ),
        Not_Observed_Cases_3Y=(
            "Recovery_Status_3Y",
            lambda x: (
                x == "Not observed within available follow-up"
            ).sum()
        ),
        No_Negative_Deviation_Cases_3Y=(
            "Recovery_Status_3Y",
            lambda x: (
                x == "No negative deviation observed"
            ).sum()
        )
    )
    .reset_index()
)

print("\nCross-event summary by product and indicator:")
display(cross_event_summary_df)


# ------------------------------------------------------------
# 6. Export consolidated results
# ------------------------------------------------------------

output_path = Path(
    "consolidated_crisis_results.xlsx"
)

with pd.ExcelWriter(
    output_path,
    engine="openpyxl"
) as writer:

    event_overview_df.to_excel(
        writer,
        sheet_name="All_12_Crisis_Events",
        index=False
    )

    consolidated_results_df.to_excel(
        writer,
        sheet_name="Consolidated_Results",
        index=False
    )

    cross_event_summary_df.to_excel(
        writer,
        sheet_name="Cross_Event_Summary",
        index=False
    )

    trajectory_coverage.to_excel(
        writer,
        sheet_name="Trajectory_Coverage",
        index=False
    )

    sensitivity_comparison.to_excel(
        writer,
        sheet_name="Sensitivity_Comparison",
        index=False
    )

print("\nStep 11.1 completed successfully.")
print("Historical events:", len(event_overview_df))
print(
    "Empirical event-product-indicator combinations:",
    len(consolidated_results_df)
)
print(f"Results exported to: {output_path.resolve()}")
```

## 11.2 Standardised Cross-Event Comparison

### 11.2.1 Objective

The objective of this step is to improve the comparability of empirical results across crisis events by applying a common post-crisis observation window.

The previous analysis includes different follow-up periods depending on the historical timing of each crisis and the available BLE data. Older events have longer observation periods than more recent events. Directly comparing summary measures based on these different periods may therefore lead to misleading conclusions.

To address this limitation, a standardised observation window is introduced.

### 11.2.2 Common Observation Window

The standardised comparison covers the event year and the following two agricultural years:

- Event year: t = 0
- Short-term follow-up: t+1
- Short-term follow-up: t+2

The resulting observation window comprises three agricultural years for each crisis event.

The comparison includes all events for which the main empirical analysis is available (EV03–EV11). If an event does not have all three observations available, it is excluded from the standardised comparison or reported separately, depending on the specific analytical measure.

### 11.2.3 Standardised Deviation Measures

For each product, indicator, and crisis event, annual deviations from the three-year pre-crisis reference level are examined.

The following measures are considered:

- Mean deviation over the common three-year observation window
- Minimum deviation within the observation window
- Maximum deviation within the observation window
- Number of available observations

The mean deviation is calculated as:

\[
\overline{D}_{0:2}
=
\frac{1}{3}
\sum_{t=0}^{2} D_t
\]

where \(D_t\) represents the percentage deviation from the pre-crisis reference level in agricultural year \(t\).

The calculation is only performed when all three annual observations are available. This ensures that all included crisis events are compared over an identical period.

### 11.2.4 Product- and Indicator-Specific Comparisons

The standardised results are compared separately for barley, potatoes, and soft wheat.

The interpretation takes account of the economic meaning of each indicator:

- Usable production reflects domestic agricultural production and may be influenced by weather and growing conditions.
- Standardised domestic use reflects domestic utilisation and is not inherently a negative or positive outcome.
- The self-sufficiency rate reflects the relationship between domestic production and domestic use. Changes are therefore interpreted in relation to both components.

The standardised comparison is descriptive and does not establish causal effects of individual crisis events.

### 11.2.5 Methodological Limitations

The common observation window improves temporal comparability but limits the analysis to short-term developments.

It does not capture medium- or long-term recovery patterns. These remain covered by the event-specific trajectory and resilience analyses in the preceding steps.

Furthermore, crisis events may overlap. Consequently, observations in the common window may reflect multiple simultaneous or subsequent developments.

The standardised comparison therefore provides an additional robustness perspective rather than replacing the full event-specific analysis.

### 11.2.6 Role in the Overall Analysis

The standardised cross-event comparison complements the main resilience analysis and the five-year sensitivity analysis.

It allows short-term deviations to be compared across events using an identical observation horizon, while preserving the longer event-specific trajectories for the assessment of recovery and persistence.

```python

# ============================================================
# STEP 11.2 – Standardised Cross-Event Comparison
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# 1. Define the common observation window
# ------------------------------------------------------------

pre_years = 3

common_event_times = [0, 1, 2]

expected_events = [
    "EV03", "EV04", "EV05", "EV06",
    "EV07", "EV08", "EV09", "EV10", "EV11"
]

# ------------------------------------------------------------
# 2. Select the main specification and common time window
# ------------------------------------------------------------

standardised_data = trajectory_df[
    (trajectory_df["Pre_Years"] == pre_years)
    & (trajectory_df["Event_Time"].isin(common_event_times))
].copy()

# Check for duplicate event-product-indicator-time observations
duplicate_check = standardised_data.duplicated(
    [
        "Event_ID",
        "Bilanz_Produkt",
        "Indicator",
        "Event_Time"
    ]
)

assert not duplicate_check.any(), (
    "Duplicate observations found in the standardised dataset."
)

print("Observations selected:", len(standardised_data))
print("Events represented:", sorted(
    standardised_data["Event_ID"].unique()
))


# ------------------------------------------------------------
# 3. Identify combinations with a complete three-year window
# ------------------------------------------------------------

group_columns = [
    "Event_ID",
    "Bilanz_Produkt",
    "Indicator"
]

coverage_df = (
    standardised_data
    .groupby(group_columns)
    .agg(
        Available_Years=("Event_Time", "nunique"),
        First_Event_Time=("Event_Time", "min"),
        Last_Event_Time=("Event_Time", "max")
    )
    .reset_index()
)

complete_combinations = coverage_df[
    (coverage_df["Available_Years"] == 3)
    & (coverage_df["First_Event_Time"] == 0)
    & (coverage_df["Last_Event_Time"] == 2)
][group_columns]

# Retain only complete combinations
complete_data = standardised_data.merge(
    complete_combinations,
    on=group_columns,
    how="inner",
    validate="many_to_one"
)

print("\nComplete three-year combinations:")
print(len(complete_combinations))

print(
    "Events included in standardised comparison:",
    sorted(complete_combinations["Event_ID"].unique())
)


# ------------------------------------------------------------
# 4. Calculate standardised deviation measures
# ------------------------------------------------------------

standardised_comparison_df = (
    complete_data
    .groupby(group_columns)
    .agg(
        Mean_Deviation_0_to_2=(
            "Deviation_From_Mean_Pct",
            "mean"
        ),
        Minimum_Deviation_0_to_2=(
            "Deviation_From_Mean_Pct",
            "min"
        ),
        Maximum_Deviation_0_to_2=(
            "Deviation_From_Mean_Pct",
            "max"
        ),
        Available_Years=("Event_Time", "nunique")
    )
    .reset_index()
)

# Add historical event information
event_metadata = pd.DataFrame(crisis_events)[
    ["Event_ID", "Event_Name", "Start_Month"]
]

standardised_comparison_df = standardised_comparison_df.merge(
    event_metadata,
    on="Event_ID",
    how="left",
    validate="many_to_one"
)

standardised_comparison_df = (
    standardised_comparison_df
    .sort_values(
        ["Start_Month", "Bilanz_Produkt", "Indicator"]
    )
    .reset_index(drop=True)
)

print("\nStandardised cross-event comparison:")
print(standardised_comparison_df.shape)

display(standardised_comparison_df)


# ------------------------------------------------------------
# 5. Create product- and indicator-specific summaries
# ------------------------------------------------------------

cross_event_standardised_summary_df = (
    standardised_comparison_df
    .groupby(
        ["Bilanz_Produkt", "Indicator"]
    )
    .agg(
        Number_of_Events=("Event_ID", "nunique"),
        Mean_Deviation_Across_Events=(
            "Mean_Deviation_0_to_2",
            "mean"
        ),
        Mean_Minimum_Deviation=(
            "Minimum_Deviation_0_to_2",
            "mean"
        ),
        Mean_Maximum_Deviation=(
            "Maximum_Deviation_0_to_2",
            "mean"
        )
    )
    .reset_index()
)

print("\nCross-event summary:")
display(cross_event_standardised_summary_df)


# ------------------------------------------------------------
# 6. Create an event coverage report
# ------------------------------------------------------------

event_coverage_df = (
    coverage_df
    .groupby("Event_ID")
    .agg(
        Complete_Combinations=(
            "Available_Years",
            lambda x: (x == 3).sum()
        ),
        Total_Combinations=("Indicator", "size"),
        Maximum_Available_Years=(
            "Available_Years",
            "max"
        ),
        Minimum_Available_Years=(
            "Available_Years",
            "min"
        )
    )
    .reset_index()
)

# Add all historically defined events, including those without
# observations in the common window
event_coverage_df = event_metadata.merge(
    event_coverage_df,
    on="Event_ID",
    how="left"
)

print("\nStandardised observation coverage by event:")
display(event_coverage_df)


# ------------------------------------------------------------
# 7. Validation
# ------------------------------------------------------------

print("\nVALIDATION")

assert standardised_comparison_df["Available_Years"].eq(3).all(), (
    "Some combinations do not contain exactly three observations."
)

assert set(
    standardised_comparison_df["Event_ID"].unique()
) == {
    "EV03", "EV04", "EV05", "EV06",
    "EV07", "EV08", "EV09", "EV10"
}, (
    "Unexpected event coverage in the standardised comparison."
)

assert len(standardised_comparison_df) == 72, (
    "Expected 72 complete event-product-indicator combinations."
)

assert not standardised_comparison_df.duplicated(
    group_columns
).any(), (
    "Duplicate combinations in standardised comparison."
)

print("All validation checks passed.")
print("Included events: 8")
print("Products: 3")
print("Indicators: 3")
print("Complete combinations: 72")


# ------------------------------------------------------------
# 8. Export results
# ------------------------------------------------------------

output_path = Path(
    "standardised_cross_event_comparison.xlsx"
)

with pd.ExcelWriter(
    output_path,
    engine="openpyxl"
) as writer:

    standardised_comparison_df.to_excel(
        writer,
        sheet_name="Standardised_Comparison",
        index=False
    )

    cross_event_standardised_summary_df.to_excel(
        writer,
        sheet_name="Cross_Event_Summary",
        index=False
    )

    event_coverage_df.to_excel(
        writer,
        sheet_name="Event_Coverage",
        index=False
    )

print("\nStep 11.2 completed successfully.")
print("Results exported to:", output_path.resolve())
```

## 12. Overall Results, Discussion and Conclusion

### 12.1 Summary of the Research Approach

This study examines the resilience of German agricultural markets for soft wheat, barley, and potatoes in response to major historical crises. The empirical analysis is based on annual agricultural supply balance data published by the German Federal Office for Agriculture and Food (BLE).

The observation period covers the agricultural years 1991/92 to 2024/25. The agricultural year runs from 1 July to 30 June, and this reporting structure is retained throughout the analysis to ensure consistency with the original data.

Three indicators are used to examine different dimensions of agricultural market developments:

- **Usable production:** measures the domestically available agricultural output.
- **Standardised domestic use:** captures the domestic utilisation of the respective agricultural product.
- **Self-sufficiency rate:** indicates the relationship between domestic production and domestic use.

The historical crisis chronology comprises twelve events. Their starting dates were established independently of the empirical market developments to avoid defining crises retrospectively based on observed changes in the indicators. Historical event dates were subsequently assigned to the corresponding BLE agricultural years.

The empirical analysis follows a comparative event-based approach. For each eligible crisis event, observed indicator values are compared with a pre-crisis reference level. The main specification uses the mean of the three preceding agricultural years. A five-year pre-crisis reference period is included as a sensitivity analysis.

Annual deviations from the pre-crisis reference level are examined over an observation window of up to ten agricultural years following the event year. The analysis distinguishes between short-term (t+1 to t+2), medium-term (t+3 to t+5), and long-term developments (t+6 to t+10).

Market resilience is assessed descriptively using several complementary measures, including the deepest negative deviation, the timing and duration of recovery, cumulative deviations, and mean annual deviations. The sensitivity analysis evaluates whether the resulting assessments are robust to alternative pre-crisis reference periods.

Finally, standardised cross-event comparisons use a common observation window of t=0 to t+2. This allows short-term developments to be compared across events using the same temporal basis.

The analysis is descriptive and does not identify causal effects. In particular, annual observations limit the identification of exact within-year responses, while overlapping crises and broader structural developments complicate the attribution of observed changes to individual events.

```python
# Step 12.3 – Linear Trend Analysis

from scipy.stats import linregress

# Define product order and English display labels
product_order = ["Weichweizen", "Gerste", "Kartoffeln"]

product_labels = {
    "Weichweizen": "Soft wheat",
    "Gerste": "Barley",
    "Kartoffeln": "Potatoes"
}

indicator_labels = {
    "Verwendbare_Erzeugung": "Usable production",
    "Inlandsverwendung_standardisiert": "Standardised domestic utilisation",
    "Selbstversorgungsgrad_in_%": "Self-sufficiency rate"
}

# Calculate linear trends for each product and indicator
trend_results = []

for product in product_order:

    product_data = analysis_df[
        analysis_df["Bilanz_Produkt"] == product
    ].sort_values("Year")

    for indicator in indicators:

        x = product_data["Year"].to_numpy()
        y = product_data[indicator].to_numpy()

        # Estimate linear trend
        result = linregress(x, y)

        trend_results.append({
            "Bilanz_Produkt": product,
            "Product": product_labels[product],
            "Indicator": indicator,
            "Indicator_Label": indicator_labels[indicator],
            "Observations": len(y),
            "Trend_Slope_Per_Year": result.slope,
            "Intercept": result.intercept,
            "R_Squared": result.rvalue ** 2,
            "P_Value": result.pvalue
        })

# Convert results into a DataFrame
trend_analysis_df = pd.DataFrame(trend_results)

# Apply the predefined product order
trend_analysis_df["Bilanz_Produkt"] = pd.Categorical(
    trend_analysis_df["Bilanz_Produkt"],
    categories=product_order,
    ordered=True
)

# Sort results
trend_analysis_df = (
    trend_analysis_df
    .sort_values(["Bilanz_Produkt", "Indicator"])
    .reset_index(drop=True)
)

# Select output columns
trend_analysis_df = trend_analysis_df[
    [
        "Product",
        "Indicator_Label",
        "Observations",
        "Trend_Slope_Per_Year",
        "R_Squared",
        "P_Value"
    ]
]

# Display results
display(trend_analysis_df.round(4))

# Validate results
assert len(trend_analysis_df) == 9
assert trend_analysis_df["Observations"].eq(34).all()
assert trend_analysis_df["Trend_Slope_Per_Year"].notna().all()

print("Linear trend analysis completed successfully.")
print("Products:", trend_analysis_df["Product"].nunique())
print("Indicators:", trend_analysis_df["Indicator_Label"].nunique())
print("Total trend estimates:", len(trend_analysis_df))
```

For barley, both usable production and standardised domestic utilisation exhibit negative trends. The self-sufficiency rate remains comparatively stable over the full period.

The trend analysis therefore provides important context for the subsequent examination of crisis-related deviations. It does not replace the event-specific analysis and does not establish causal effects of individual crises.

```python
# Step 12.4 – Autocorrelation and Time-Series Diagnostics

from scipy.stats import linregress, chi2
import numpy as np
import pandas as pd

# Define product order and English display labels
product_order = ["Weichweizen", "Gerste", "Kartoffeln"]

product_labels = {
    "Weichweizen": "Soft wheat",
    "Gerste": "Barley",
    "Kartoffeln": "Potatoes"
}

indicator_labels = {
    "Verwendbare_Erzeugung": "Usable production",
    "Inlandsverwendung_standardisiert": "Standardised domestic utilisation",
    "Selbstversorgungsgrad_in_%": "Self-sufficiency rate"
}

diagnostic_results = []

for product in product_order:

    product_data = (
        analysis_df[
            analysis_df["Bilanz_Produkt"] == product
        ]
        .sort_values("Year")
    )

    for indicator in indicators:

        data = product_data[["Year", indicator]].dropna()

        x = data["Year"].to_numpy(dtype=float)
        y = data[indicator].to_numpy(dtype=float)

        # Estimate the linear trend
        trend_result = linregress(x, y)

        # Calculate residuals from the linear trend
        fitted_values = (
            trend_result.intercept
            + trend_result.slope * x
        )

        residuals = y - fitted_values

        # Durbin-Watson statistic
        dw_statistic = (
            np.sum(np.diff(residuals) ** 2)
            / np.sum(residuals ** 2)
        )

        # Lag-1 autocorrelation
        lag1_autocorrelation = np.corrcoef(
            residuals[:-1],
            residuals[1:]
        )[0, 1]

        # Ljung-Box test for lags 1 and 2
        n = len(residuals)
        max_lag = 2

        autocorrelations = [
            np.corrcoef(
                residuals[:-lag],
                residuals[lag:]
            )[0, 1]
            for lag in range(1, max_lag + 1)
        ]

        ljung_box_statistic = n * (n + 2) * sum(
            autocorrelations[lag - 1] ** 2 / (n - lag)
            for lag in range(1, max_lag + 1)
        )

        # Approximate p-value with 2 degrees of freedom
        ljung_box_p_value = chi2.sf(
            ljung_box_statistic,
            df=max_lag
        )

        diagnostic_results.append({
            "Product": product_labels[product],
            "Indicator": indicator_labels[indicator],
            "Observations": n,
            "Durbin_Watson": dw_statistic,
            "Lag_1_Autocorrelation": lag1_autocorrelation,
            "Ljung_Box_Statistic_Lag_2": ljung_box_statistic,
            "Ljung_Box_P_Value_Lag_2": ljung_box_p_value
        })

# Create results DataFrame
time_series_diagnostics_df = pd.DataFrame(
    diagnostic_results
)

# Display results
display(time_series_diagnostics_df.round(4))

# Validate results
assert len(time_series_diagnostics_df) == 9
assert time_series_diagnostics_df["Observations"].eq(34).all()

print("Time-series diagnostics completed successfully.")
print("Product-indicator combinations:", len(time_series_diagnostics_df))
```

### 12.4.2 Summary of Autocorrelation Diagnostics

The autocorrelation diagnostics indicate positive first-order dependence in several product-indicator combinations. This pattern is particularly pronounced for usable production and standardised domestic utilisation in the soft wheat market, as well as for the self-sufficiency rate in the potato market.

The barley indicators generally exhibit weaker to moderate positive autocorrelation. Potato usable production shows little first-order autocorrelation.

These findings suggest that deviations from the estimated linear trends may persist across consecutive agricultural years. Consequently, observations cannot necessarily be treated as fully independent, and conventional regression significance estimates should be interpreted cautiously.

The Durbin–Watson statistic and lag-1 autocorrelation are used as descriptive diagnostic measures. The Ljung–Box results are not interpreted as formal significance tests because of the short time series and the approximate reference distribution used after trend estimation.

Overall, the diagnostics provide additional context for interpreting long-term trends and crisis-related deviations. They do not establish causal relationships or identify the source of the observed temporal dependence.

```python
# Step 12.5 – Structural Break and Outlier Analysis

import numpy as np
import pandas as pd
from scipy.stats import linregress

# Define product order and English display labels
product_order = ["Weichweizen", "Gerste", "Kartoffeln"]

product_labels = {
    "Weichweizen": "Soft wheat",
    "Gerste": "Barley",
    "Kartoffeln": "Potatoes"
}

indicator_labels = {
    "Verwendbare_Erzeugung": "Usable production",
    "Inlandsverwendung_standardisiert": "Standardised domestic utilisation",
    "Selbstversorgungsgrad_in_%": "Self-sufficiency rate"
}

# Minimum number of observations on each side of a candidate break
min_segment_size = 8

# Robust outlier threshold
outlier_threshold = 3.5

outlier_results = []
break_results = []

for product in product_order:

    product_data = (
        analysis_df[
            analysis_df["Bilanz_Produkt"] == product
        ]
        .sort_values("Year")
    )

    for indicator in indicators:

        data = (
            product_data[["Year", indicator]]
            .dropna()
            .sort_values("Year")
        )

        x = data["Year"].to_numpy(dtype=float)
        y = data[indicator].to_numpy(dtype=float)

        n = len(y)

        # --------------------------------------------------
        # 1. Outlier screening based on linear trend residuals
        # --------------------------------------------------

        full_trend = linregress(x, y)

        fitted_values = (
            full_trend.intercept
            + full_trend.slope * x
        )

        residuals = y - fitted_values

        residual_median = np.median(residuals)
        absolute_deviations = np.abs(
            residuals - residual_median
        )

        mad = np.median(absolute_deviations)

        if mad > 0:
            robust_scores = (
                0.6745
                * (residuals - residual_median)
                / mad
            )
        else:
            robust_scores = np.full(n, np.nan)

        outlier_mask = (
            np.abs(robust_scores) > outlier_threshold
        )

        for i in range(n):
            outlier_results.append({
                "Product": product_labels[product],
                "Indicator": indicator_labels[indicator],
                "Year": int(x[i]),
                "Agricultural_Year": data.iloc[i]["Year"],
                "Observed_Value": y[i],
                "Trend_Residual": residuals[i],
                "Robust_Score": robust_scores[i],
                "Potential_Outlier": bool(outlier_mask[i])
            })

        # --------------------------------------------------
        # 2. Exploratory structural break screening
        # --------------------------------------------------

        # Baseline model: one linear trend over the full period
        baseline_sse = np.sum(residuals ** 2)

        best_break = None
        best_sse = np.inf
        best_slope_before = np.nan
        best_slope_after = np.nan

        # Test candidate break positions
        for break_index in range(
            min_segment_size,
            n - min_segment_size + 1
        ):

            x_before = x[:break_index]
            y_before = y[:break_index]

            x_after = x[break_index:]
            y_after = y[break_index:]

            # Fit separate linear trends
            trend_before = linregress(
                x_before,
                y_before
            )

            trend_after = linregress(
                x_after,
                y_after
            )

            fitted_before = (
                trend_before.intercept
                + trend_before.slope * x_before
            )

            fitted_after = (
                trend_after.intercept
                + trend_after.slope * x_after
            )

            segmented_sse = (
                np.sum((y_before - fitted_before) ** 2)
                + np.sum((y_after - fitted_after) ** 2)
            )

            if segmented_sse < best_sse:
                best_sse = segmented_sse
                best_break = int(x[break_index])
                best_slope_before = trend_before.slope
                best_slope_after = trend_after.slope

        # Calculate the relative reduction in residual SSE
        if baseline_sse > 0:
            sse_reduction_pct = (
                (baseline_sse - best_sse)
                / baseline_sse
                * 100
            )
        else:
            sse_reduction_pct = np.nan

        break_results.append({
            "Product": product_labels[product],
            "Indicator": indicator_labels[indicator],
            "Observations": n,
            "Baseline_Slope": full_trend.slope,
            "Candidate_Break_Year": best_break,
            "Slope_Before_Break": best_slope_before,
            "Slope_After_Break": best_slope_after,
            "Baseline_SSE": baseline_sse,
            "Segmented_SSE": best_sse,
            "SSE_Reduction_Pct": sse_reduction_pct
        })

# Create result DataFrames
outlier_diagnostics_df = pd.DataFrame(outlier_results)

structural_break_screening_df = pd.DataFrame(
    break_results
)

# Display only potential outliers
potential_outliers_df = (
    outlier_diagnostics_df[
        outlier_diagnostics_df["Potential_Outlier"]
    ]
    .sort_values(["Product", "Indicator", "Year"])
    .reset_index(drop=True)
)

print("Potential outliers:")
display(potential_outliers_df.round(3))

print("Structural break screening:")
display(
    structural_break_screening_df.round(3)
)

# Summary of outlier screening
outlier_summary_df = (
    outlier_diagnostics_df
    .groupby(["Product", "Indicator"])
    .agg(
        Observations=("Year", "count"),
        Potential_Outliers=("Potential_Outlier", "sum")
    )
    .reset_index()
)

print("Outlier screening summary:")
display(outlier_summary_df)

# Validate results
assert len(structural_break_screening_df) == 9
assert len(outlier_diagnostics_df) == 9 * 34
assert structural_break_screening_df["Observations"].eq(34).all()

print("Structural break and outlier screening completed successfully.")
print("Product-indicator combinations:", len(structural_break_screening_df))
print("Potential outliers identified:", len(potential_outliers_df))
```

### 12.5.2 Summary of Outlier and Structural Break Screening

The outlier screening identified no potential outliers among the 306 observations across the nine product-indicator combinations. No observation exceeded the predefined robust threshold of 3.5 MAD from the estimated linear trend.

The exploratory structural break screening identified candidate break years for all nine time series. Several indicators displayed substantial improvements in model fit when the observation period was divided into two separate linear trend segments.

Particularly pronounced improvements were found for potato self-sufficiency, soft wheat usable production, soft wheat domestic utilisation, and potato domestic utilisation. In some cases, the estimated trend slopes changed direction between the two segments.

These findings suggest that a single linear trend may not adequately describe the entire observation period for every indicator. However, the candidate break years are exploratory results rather than statistically confirmed structural breaks. The reduction in residual sum of squares does not account for the search across multiple candidate years and must not be interpreted as proof of a structural change.

No observations were removed or modified. The identified candidate years are retained as points for further contextual interpretation alongside the historical crisis chronology. No causal relationship between candidate break years and individual crisis events is inferred.

## 12.6 Robustness of the Crisis Impact Analysis

### 12.6.1 Objective

This section evaluates the robustness of the empirical crisis impact analysis by comparing alternative methodological specifications and assessing limitations arising from data availability, overlapping events, and incomplete observation periods.

The objective is to determine whether the main findings depend strongly on the selected reference period or analytical approach. The robustness assessment builds on the analyses already conducted in Steps 9 to 11.

### 12.6.2 Sensitivity to the Pre-Crisis Reference Period

The main crisis impact analysis uses a three-year pre-crisis reference period (t−3 to t−1). A five-year reference period (t−5 to t−1) was additionally applied as a sensitivity specification.

The comparison showed that the resulting resilience measures are not identical across the two specifications. Of the 72 paired comparisons, 62 (86.1%) produced the same recovery classification, while 10 (13.9%) differed.

This indicates that the recovery classification is relatively consistent in most cases, but remains sensitive to the reference period for certain product-indicator combinations.

The three-year reference period is therefore retained as the main specification, while the five-year reference period serves as a robustness check.

### 12.6.3 Sensitivity to the Reference Model

The crisis impact analysis considers deviations from both the pre-crisis mean and the extrapolated pre-crisis linear trend.

The pre-crisis mean provides a reference for the average level before the event. The linear trend additionally accounts for the direction of development during the reference period.

Differences between these two approaches indicate that the interpretation of post-crisis developments can depend on whether the reference is based on a constant average level or an extrapolated trend.

Neither specification establishes a causal crisis effect. Both provide alternative descriptive reference points for assessing subsequent market developments.

```python
# Step 12.6 – Robustness of the Crisis Impact Analysis

# --------------------------------------------------
# 1. Sensitivity to the pre-crisis reference period
# --------------------------------------------------

reference_period_comparison_df = sensitivity_comparison.copy()

reference_period_summary_df = (
    reference_period_comparison_df
    .groupby(["Bilanz_Produkt", "Indicator"])
    .agg(
        Paired_Comparisons=("Event_ID", "count"),
        Mean_Change_Deepest_Deviation=(
            "Deepest_Negative_Deviation_Pct_Difference_5Y_minus_3Y",
            "mean"
        ),
        Mean_Change_Cumulative_Deviation=(
            "Cumulative_Deviation_Pct_Difference_5Y_minus_3Y",
            "mean"
        ),
        Mean_Change_Annual_Deviation=(
            "Mean_Annual_Deviation_Pct_Difference_5Y_minus_3Y",
            "mean"
        )
    )
    .reset_index()
)

print("Reference period sensitivity:")
display(reference_period_summary_df.round(3))


# --------------------------------------------------
# 2. Agreement of recovery classifications
# --------------------------------------------------

recovery_agreement_df = sensitivity_comparison[
    [
        "Event_ID",
        "Bilanz_Produkt",
        "Indicator",
        "Recovery_Status_3Y",
        "Recovery_Status_5Y",
        "Recovery_Status_Consistent"
    ]
].copy()

total_comparisons = len(recovery_agreement_df)

matching_comparisons = recovery_agreement_df[
    "Recovery_Status_Consistent"
].sum()

different_comparisons = (
    total_comparisons - matching_comparisons
)

agreement_rate = (
    matching_comparisons / total_comparisons * 100
    if total_comparisons > 0
    else np.nan
)

recovery_agreement_summary_df = pd.DataFrame({
    "Measure": [
        "Total paired comparisons",
        "Matching recovery classifications",
        "Different recovery classifications",
        "Agreement rate (%)"
    ],
    "Value": [
        total_comparisons,
        matching_comparisons,
        different_comparisons,
        agreement_rate
    ]
})

print("Recovery classification agreement:")
display(recovery_agreement_summary_df.round(2))


# --------------------------------------------------
# 3. Availability of post-crisis observations
# --------------------------------------------------

event_coverage_summary_df = event_coverage_df.copy()

print("Event observation coverage:")
display(event_coverage_summary_df)


# --------------------------------------------------
# 4. Overlapping crisis event windows
# --------------------------------------------------

# Map all event starting months to BLE agricultural years
overlap_records = []

for event in crisis_events:
    start_month = pd.Period(event["Start_Month"], freq="M")
    event_year = month_to_agricultural_year(start_month)

    overlap_records.append({
        "Event_ID": event["Event_ID"],
        "Event_Name": event["Event_Name"],
        "Start_Month": event["Start_Month"],
        "BLE_Event_Year": event_year
    })

event_start_overview_df = pd.DataFrame(overlap_records)

# Identify events starting in the same BLE agricultural year
overlapping_start_years_df = (
    event_start_overview_df
    .groupby("BLE_Event_Year")
    .agg(
        Number_of_Events=("Event_ID", "count"),
        Event_IDs=("Event_ID", lambda x: ", ".join(x))
    )
    .reset_index()
)

overlapping_start_years_df = overlapping_start_years_df[
    overlapping_start_years_df["Number_of_Events"] > 1
]

print("Events starting in the same BLE agricultural year:")
display(overlapping_start_years_df)


# --------------------------------------------------
# 5. Overall robustness summary
# --------------------------------------------------

overall_robustness_df = pd.DataFrame({
    "Robustness_Check": [
        "Three-year versus five-year reference period",
        "Recovery classification agreement",
        "Observation coverage",
        "Overlapping crisis event starts"
    ],
    "Result": [
        f"{total_comparisons} paired comparisons",
        f"{matching_comparisons}/{total_comparisons} matching classifications",
        "Follow-up coverage varies across events",
        f"{len(overlapping_start_years_df)} BLE years with multiple event starts"
    ],
    "Interpretation": [
        "Sensitivity to the selected reference period",
        "Consistency between recovery classifications",
        "Unequal follow-up periods limit comparability",
        "Some event windows share agricultural-year observations"
    ]
})

print("Overall robustness assessment:")
display(overall_robustness_df)


# --------------------------------------------------
# 6. Validation
# --------------------------------------------------

assert total_comparisons == 72
assert matching_comparisons == 62
assert different_comparisons == 10

print("Robustness assessment completed successfully.")
print("Paired sensitivity comparisons:", total_comparisons)
print("Matching recovery classifications:", matching_comparisons)
print("Different recovery classifications:", different_comparisons)
```

### 12.6.8 Results of the Robustness Checks

The robustness analysis compared crisis impact estimates based on three-year and five-year pre-crisis reference periods. A total of 72 paired event-product-indicator comparisons were available for this assessment.

The recovery classifications were consistent in 62 of the 72 comparisons, corresponding to an agreement rate of 86.11%. In the remaining 10 cases, the recovery classification differed depending on the selected reference period. These differences indicate that recovery assessments are generally, but not entirely, robust to the length of the pre-crisis reference period.

The sensitivity results also show that changes in the reference period affect the estimated depth of negative deviations and cumulative deviations to varying degrees across products and indicators. The differences are not uniform, which suggests that the influence of the reference-period specification is context-dependent.

Observation coverage is an additional limitation. The available follow-up ranges from eleven agricultural years for EV03–EV05 to only two years for EV11. Consequently, recovery assessments for more recent events are provisional and cannot be compared directly with those of earlier events with longer observation periods.

A further limitation concerns overlapping crisis events. EV08, EV09 and EV10 all begin in the 2021/22 BLE agricultural year. Their event-time trajectories therefore share annual observations, making it difficult to distinguish their individual contributions to observed changes in agricultural market indicators.

Overall, the robustness checks support the use of the three-year pre-crisis reference period as the main specification, while demonstrating that some recovery classifications depend on the reference-period choice. The results should be interpreted as descriptive evidence of temporal deviations rather than as causal estimates of individual crisis effects.
